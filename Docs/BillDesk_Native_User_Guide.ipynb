### 1.2 What BillDesk does

*Screen: All*


BillDesk is a desktop program for the billing counter and the back office of a wholesale fruit and vegetable business. One program, one database, one set of numbers.

| You do this… | …and BillDesk automatically |
|---|---|
| Make a **bill** | reduces stock, adds to the customer's balance, records the sale and its cost in the accounts |
| Receive **payment** | reduces the customer's balance, records cash or bank in the accounts, closes the bill |
| Enter a **supplier bill** | adds what you owe the supplier (and TDS if it applies) |
| **Return goods** | credits the customer, puts good stock back, reverses the sale in the accounts |
| **Void** a bill | puts the stock back, reverses the balance and the accounts |
| Take a **customer order** | lists it in the order matrix so you know how much to buy, and turns it into a bill with one click |

GST is not used (fresh fruit and vegetables are exempt), so there is no tax set-up. A GSTIN can still be stored for a customer or supplier if you want it on file.

**Who sees what.** Menus depend on your role. Cashiers normally see Billing, Orders and Bill History; managers also see Masters, Reports and Accounts; administrators see everything including Settings. If a menu is missing, your role does not include it: ask the administrator ([Users and roles](#users)).



**See also:** [The dashboard](#dashboard) · [Signing in](#signing-in)

### 1.3 Signing in

*Screen: Login*


1. Double-click **BillDesk** on the desktop or in the Start menu. The first start takes a few seconds because BillDesk starts its own database.
2. Type your **username** and **password**.
3. Press **Enter** or click **Login**.



* **First time ever:** the installer creates one administrator named `admin` with a **random password shown once**. Write it down, then create the other users in [Users and roles](#users).
* **Forgot the administrator password?** Ask your IT person to run `python scripts\reset_password.py --user admin`.
* Only **one BillDesk window** can be open on a computer; if you start it twice, the second copy tells you.
* To sign out press **F12** (you are asked to confirm).

## Chapter 2. Billing at the counter

<a id="new-bill-overview"></a>

### 2.1 Making a bill

*Screen: New Bill*


Open **File → New Bill** or press **F2**.

1. Press **F5** and choose the customer (or leave **Cash** for a walk-in).
2. The cursor moves to the **Date**; accept today or pick another day from the calendar.
3. Type each item's **code** and press **Enter**; type the **quantity**, **Enter**; the unit and the rate follow.
4. Press **F2** to take payment (or **F3** to save **and** print).
5. Check the amount received and press **Post Payment**.



| Part of the screen | What it is |
|---|---|
| **Billing Company** | which of your companies the invoice is issued under |
| **Customer (F5)** | who the bill is for ([choosing the customer](#bill-customer)) |
| **Date** | the bill date, with a drop-down calendar ([dates](#bill-date)) |
| **The grid** | one row per item: Code, Item Description, Qty, Unit, Rate, Amount, and a red **✕** to clear the row |
| **Line under the grid** | stock and rate of the item you just entered ([the hint line](#bill-items)) |
| **Customer (Delivery) / Bill To** | where the goods go and who is invoiced |
| **Internal Notes** | for your own use, not printed |
| **Total** | the bill total |
| **Park (F6) / Parked (F7)** | put a bill on hold, or bring one back |





**See also:** [Choosing the customer](#bill-customer) · [The bill date and the calendar](#bill-date) · [Entering items](#bill-items) · [Taking payment](#bill-payment)

### 2.2 Choosing the customer

*Screen: New Bill*


Press **F5** (or click the customer box).

1. Type part of the customer's name or phone number: the list narrows as you type.
2. Press **↓** or **↑** to move, **Enter** to choose. (Esc closes the list; a click also chooses.)
3. Choose **Cash Customer** for a walk-in sale.



* The customer's address, their "bill to" name and their company are filled in for you, and any **fixed contract rates** for that customer are used.
* The same dialog is used on [New customer order](#order-new), so it works the same everywhere. Orders need a real customer (no Cash option there).
* A customer who is not in the list must be added first in the [Customer Master](#customers).



**See also:** [Making a bill](#new-bill-overview) · [The bill date and the calendar](#bill-date) · [Customers](#customers)

### 2.3 The bill date and the calendar

*Screen: New Bill*


Right after you choose the customer the cursor goes to **Date** and a **calendar drops down** under it.

1. Click a day, or use the arrow keys and press **Enter**. **Today** jumps to today.
2. Or type a date (`13/10/2026`, `13-10-26`) and press **Enter**.
3. The cursor moves on to the first empty row of the grid.



* The date you choose **is the date of the bill** and its invoice number follows that day (`20261005-0003` for 5 October).
* A bill cannot be dated in the future.
* More about date boxes everywhere in the program: [Dates and the calendar](#dates-everywhere).



**See also:** [Dates and the calendar](#dates-everywhere) · [Entering items](#bill-items)

<a id="bill-items"></a>

### 2.4 Entering items

*Screen: New Bill*


Work row by row with the keyboard: **Code → Qty → Unit → Rate → next row**, pressing **Enter** each time.

1. Type the item **code** (for example `111`) or a part of its **name** (`tom`), press **Enter**. The description, unit and rate appear and the cursor goes to **Qty** (1 is filled in).
2. Type the **quantity**, press **Enter**. The cursor passes through **Unit** to **Rate**.
3. Change the **rate** if you need to; press **Enter** to go to the next row.



* **The hint line** under the grid tells you the item, its stock ("No stock recorded" if stock is not tracked for it) and the rate and where it came from: *fixed rate for <customer>* or *item rate*.
* **Several items fit what you typed?** A list opens: **↑ / ↓** to move, **Enter** to choose, **Esc** to cancel. BillDesk never picks the first match for you.



* An item typed in a lower row **moves up** to the first empty row, so there are never gaps.
* The red **✕** clears a row and the rows below move up.
* You can sell more than the stock shows; BillDesk warns but does not stop you (your administrator can change this).
* Quantity and rate must be numbers; a bill cannot have a zero quantity or a negative rate.



**See also:** [The same item twice](#bill-duplicate) · [Taking payment](#bill-payment) · [Fixed contract rates](#fixed-rates)

### 2.6 Taking payment

*Screen: New Bill*


1. Press **F2** (save) or **F3** (save and print).
2. The amount received is filled in with the whole total. Change it if the customer pays part.
3. Choose the **payment method** and, for UPI or a cheque, the **reference**.
4. Press **Enter** or **Post Payment**.



| Box | Meaning |
|---|---|
| **Outstanding / Unpaid bills** | what this customer already owes before this bill |
| **Amount Received** | what the customer pays now; anything not paid stays on their account |
| **Payment Date** | shown for information: a counter payment is recorded on the bill's date |
| **Payment Method** | Cash, UPI, Cheque, NEFT/RTGS, or **Credit/Due** (nothing paid now) |
| **Reference / UTR** | UPI or cheque number (optional) |

* You cannot receive more than the bill total.
* After posting, the sheet is cleared for the next bill. With **F3** the invoice opens for printing.
* Later payments are recorded as a [customer receipt](#receipt).

### 2.7 Park a bill and recall it

*Screen: New Bill*

1. Press **F6** (or **Park**) with the items entered. The sheet clears.

3. Press **F7** (or **Parked (n)**), click the parked bill to bring it back. If another bill is on the sheet you are asked before it is replaced.



Parked bills are kept on disk: they survive closing BillDesk or a power cut.

### 3.4 Void a bill

*Screen: Bill History*


A bill made by mistake is **voided**, not deleted, so the record and the invoice number stay.

1. Open [Bill History](#history) and select the bill.
2. Press **Void Bill**.
3. Read what will happen and confirm.

Voiding: reverses the customer's balance, puts the stock back, reverses the crates and the accounts, and marks the bill **void**. Payments that had been allocated to it become the customer's **advance** (credit).

* A bill that has goods returned against it **cannot be voided**: [cancel the return](#cancel-return) first.
* A voided bill cannot be voided twice.
* To take back only some goods, use [Return goods](#return-goods), not void.



**See also:** [Return goods (credit note)](#return-goods) · [Bill History](#history)

<a id="return-goods"></a>

### 3.5 Return goods (credit note)

*Screen: Bill History*


A customer brings goods back.

1. In [Bill History](#history) select the invoice they were bought on.
2. Press **Return Goods**.
3. Type the **Return qty** for each item. Tick **Spoiled?** if the goods are rotten and cannot be sold again.
4. Write a **reason** (optional) and, for a walk-in customer, choose how you refund (Cash, UPI, Bank).
5. Press **Save return**. You are offered the **credit note** to print.



What happens:

| | Good goods | Spoiled goods |
|---|---|---|
| Customer is credited at the rate billed | yes | yes |
| Goes back into stock | yes, at the cost it was sold at | no |
| Sales are reduced in the accounts | yes | yes |

* The credit first reduces **what the customer still owes on that invoice**; anything more stays as **credit on their account** (shown as *Cr* in the customer list).
* You cannot return more than was billed minus what was already returned.
* A void invoice cannot have goods returned.
* A walk-in customer is refunded from the cash drawer or bank.



**See also:** [Credit notes](#credit-notes) · [Cancel a return](#cancel-return) · [Void a bill](#history-void) · [Customer statement](#statement)

### 3.6 Credit notes

*Screen: Bill History*


**Bill History → Credit Notes** lists the returns made on the selected invoice, each with its number, date and amount.



Select one and press **Print / Save PDF**. The credit note shows each item, whether it was good or spoiled, the total in words and how the credit was applied.





**See also:** [Return goods (credit note)](#return-goods) · [Cancel a return](#cancel-return)

<a id="cancel-return"></a>

### 3.7 Cancel a return

*Screen: Bill History, Procurement*


A return entered by mistake can be cancelled.

1. **Bill History → Credit Notes**, select the return.
2. Press **Cancel this return** and confirm.

BillDesk puts back the customer's balance, the invoice's balance due, the stock and the accounts exactly as they were; the credit note stays on file, marked **cancelled**, and no longer counts anywhere. The whole quantity can be returned again afterwards.

* A return can be cancelled once only.
* Returns made by the older program cannot be cancelled here.
* A supplier return (debit note) is cancelled from **Procurement → Returns to Suppliers → Cancel Debit Note** ([purchase return](#purchase-return)).



**See also:** [Return goods (credit note)](#return-goods) · [Return goods to a supplier (debit note)](#purchase-return) · [Credit notes](#credit-notes)

### 4.2 New customer order

*Screen: New Order*


Press **+ Create New Order** on the Orders screen, or **File → New Customer Order**.

2. The cursor goes to the **order date**, then the **delivery date**, then the first empty row of the items.

4. Enter **Crates out / in** if returnable crates go with the order.
5. Press **F3** to save (or **F10** to save and print).



| Key | Does |

### 4.4 Smart Importer (photo, scan or text)

*Screen: New Order*


Customers send orders as photos or messages. **F8** on the order form reads them for you.

1. Press **F8**.
2. Press **Upload Image…** and choose the photo or scan (PNG, JPG, BMP, TIFF, WebP).
3. Wait a few seconds: the picture shows on the left and what was read on the right.
4. Check the **customer**, the **date** and every row. A row marked **? (choose item)** could not be matched: **double-click** it to pick the item, correct the quantity or unit, or remove the line.
5. Press **Import to Grid**. The items go into the order; check the quantities, then save with **F3**.

* The date and the customer at the top of the note are picked up too.
* **Printed or typed** orders read well. **Handwriting is hard for this kind of reader**: if nothing can be read the window says so instead of guessing. Use a straight, bright photo, or type the lines with **Paste text instead**.
* The same item twice in one order is added together.

<a id="purchase-return"></a>

### 6.8 Return goods to a supplier (debit note)

*Screen: Procurement*


1. **Procurement → Vendor Purchase Bills**, select the bill the goods came on.
2. Press **↩ Return Goods to Supplier**.
3. Type the quantity of each item, the reason, and press **Save return**. You can print the **debit note**.



* The value is at the bill's rate. If **TDS** was deducted, the same share comes back, so what you owe falls by the **net** amount.
* It reduces what you still owe on that bill first; any excess is **credit with the supplier**.
* Stock goes down only if the goods had been received into stock (the bill has a goods receipt).
* The **Returns to Suppliers** tab lists them; **Print Debit Note** and **Cancel Debit Note** are there.

| *"The bill date cannot be in the future"* | the date box holds a later day | pick today or an earlier day |
| *"… is not a valid date. Use DD/MM/YYYY"* | text in a date box can't be read | pick from the calendar |
| *"The To date cannot be before …"* | To is earlier than From | change one of them |
| *"Amount received (…) is more than the bill total"* | more entered than the bill | type the bill amount or less |
| *"Item … not found"* / *"No item matches"* | wrong code | check the code in the Item Master |
| *"… must be a number"* | letters or symbols in a number box | type digits only (decimals allowed) |
| *"An item / customer named … already exists"* | the name is taken | use another name, or edit the existing one |
| *"Vendor invoice … is already recorded as PUR-…"* | the same supplier invoice was entered before | do not enter it again |
| *"Payment amount … exceeds purchase balance due"* | paying a supplier more than owed | pay the balance due or less |
| *"Waste quantity … exceeds stock on hand"* | more than the tracked stock | check the quantity |
| *"Invoice … has goods returned against it, so it cannot be voided"* | a return exists | [cancel the return](#cancel-return) first, then void |
| *"only n can still be returned"* | more than billed minus earlier returns | reduce the return quantity |
| *"Return … was made by the older BillDesk and cannot be cancelled here"* | old returns lack detail | leave it; it is already reflected in the balance |
| *"The entry does not balance"* | journal debit ≠ credit | make the two amounts equal |
| *"You already have an open cashier session"* | a drawer is open | close it first |
| *"Access Denied"* | your role does not include that screen | ask the administrator |
| *"Tesseract OCR engine is not installed"* | the image reader is missing | use **Paste text instead**, or ask IT to reinstall |
| *"The text in this image could not be read reliably"* | handwriting or a blurry photo | take a straighter, brighter photo or type the lines |
| Integrity check shows **Warning** | an older-data difference | open the line; ask your accountant if unsure |
| Cannot start: database error | the database did not start | restart the computer, then BillDesk; call IT if it repeats |

### 10.6 Glossary

*Screen: All*


The words used in BillDesk and in this guide, in plain language.

| Word | Meaning |
|---|---|
| **Bill / Invoice** | the document you give the customer for goods sold |
| **Challan (DC)** | the same bill without prices, for delivery |
| **Void** | cancel a whole bill, keeping its record |
| **Credit note** | document for goods a customer returned (their balance goes down) |
| **Debit note** | document for goods you send back to a supplier |
| **Dr / Cr** | Dr: the customer owes you; Cr: you hold their money |
| **Advance** | money received before a bill or more than was owed |
| **Receivables** | money customers owe you |
| **Payables** | money you owe suppliers |
| **PO / GRN** | purchase order / goods receipt note (what actually arrived) |
| **TDS** | tax deducted from a supplier's payment |
| **FIFO** | oldest bill settled first |
| **BRS** | bank reconciliation statement |
| **Trial balance** | list of all accounts; debits must equal credits |
| **Float** | cash put in the drawer at the start of the day |
| **Legacy bill** | a bill made by the older program (no per-bill payment record) |
| **Parked bill** | a bill put on hold |

```python
# Search this guide in plain words. Run this cell once, then use ask("...") in any cell.
# (The search engine below is the same one BillDesk's own Help screen (F9) uses.)
"""The built-in help: topics from the user guide, searched in plain words ("how do I cancel a bill?") with step-by-step guidance.

The topics live in app/assets/help/guide.json, generated from Docs/BillDesk_Native_User_Guide.md by scripts/build_user_guide.py,
so the printed guide and the help screen can never disagree."""
from __future__ import annotations

import difflib
import json
import re
from dataclasses import dataclass, field
from pathlib import Path
from typing import Any, Dict, List, Optional

HELP_DIR = Path('.')

# words people use for the same thing here; every group is treated as one word when searching
SYNONYMS = [
    {"bill", "invoice", "sale", "sales", "billing", "receipt-bill"},
    {"customer", "buyer", "party", "client", "hotel"},
    {"supplier", "vendor", "farmer", "seller"},
    {"return", "returned", "credit-note", "creditnote", "sales-return", "give-back", "bring-back"},
    {"debit-note", "debitnote", "purchase-return"},
    {"cancel", "void", "delete", "remove", "undo", "reverse", "cancelled", "voided"},
    {"pay", "payment", "paid", "receipt", "collect", "collection", "received", "cash", "upi"},
    {"owe", "owes", "due", "outstanding", "balance", "receivable", "receivables", "pending-amount", "unpaid"},
    {"backup", "back-up", "restore", "copy", "safe"},
    {"password", "login", "log-in", "signin", "sign-in", "username", "locked"},
    {"print", "printing", "pdf", "preview", "download", "save-pdf"},
    {"date", "calendar", "day", "when"},
    {"stock", "inventory", "quantity", "qty", "godown"},
    {"waste", "spoilage", "spoiled", "rotten", "damaged", "wastage"},
    {"order", "orders", "enquiry"},
    {"image", "photo", "picture", "ocr", "scan", "whatsapp", "handwritten", "tamil", "translate"},
    {"report", "reports", "daybook", "statement", "ledger", "summary"},
    {"user", "users", "staff", "role", "roles", "access", "permission", "permissions"},
    {"company", "logo", "address", "gstin", "letterhead"},
    {"shortcut", "shortcuts", "key", "keys", "keyboard", "hotkey", "f1", "f2", "f3", "f5"},
    {"error", "problem", "wrong", "not-working", "fails", "failed", "message", "issue", "trouble"},
    {"bank", "banking", "brs", "reconciliation", "cheque", "neft", "ifsc"},
    {"journal", "accounting", "trial", "balance-sheet", "profit", "loss", "p&l"},
    {"crate", "crates", "jali", "returnable"},
    {"park", "parked", "hold", "recall", "pause"},
]
STOP = {"the", "a", "an", "to", "of", "in", "on", "for", "i", "how", "do", "does", "can", "my", "is", "it", "and", "or", "with", "what", "where",
        "want", "need", "should", "when", "from", "this", "that", "me", "we", "you", "please", "there", "are", "be", "by", "at", "get", "make", "use"}


def _stem(w: str) -> str:
    for suf in ("ing", "ies", "es", "ed", "s"):
        if len(w) > len(suf) + 2 and w.endswith(suf):
            return w[: -len(suf)] + ("y" if suf == "ies" else "")
    return w


_GROUP: Dict[str, int] = {}
for _i, _g in enumerate(SYNONYMS):
    for _w in _g:
        _GROUP.setdefault(_stem(_w), _i)


def tokens(text: str) -> List[str]:
    words = re.findall(r"[a-z0-9&]+(?:-[a-z0-9]+)*", (text or "").lower())
    out = []
    for w in words:
        if w in STOP:
            continue
        out.append(_stem(w))
        if "-" in w:
            out += [_stem(p) for p in w.split("-") if p not in STOP and len(p) > 1]
    return out


def _canon(tok: str) -> str:
    g = _GROUP.get(tok)
    return f"~{g}" if g is not None else tok


@dataclass
class Hit:
    topic: Dict[str, Any]
    score: float
    snippet: str = ""
    why: str = ""
    steps: List[str] = field(default_factory=list)


class HelpService:
    def __init__(self, data: Optional[Dict[str, Any]] = None, path: Optional[Path] = None):
        if data is None:
            p = path or (HELP_DIR / "guide.json")
            data = json.loads(Path(p).read_text(encoding="utf-8")) if Path(p).exists() else {"title": "User Guide", "chapters": [], "topics": {}, "intents": [], "faq": []}
        self.data = data
        self.topics: Dict[str, Dict[str, Any]] = data.get("topics", {})
        self.chapters: List[Dict[str, Any]] = data.get("chapters", [])
        self.intents: List[Dict[str, str]] = data.get("intents", [])
        self.faq: List[Dict[str, str]] = data.get("faq", [])
        self.context: Dict[str, str] = data.get("context", {})
        self._index = {tid: self._prepare(t) for tid, t in self.topics.items()}
        self._vocab = sorted({w for ix in self._index.values() for w in ix["all"]} | set(_GROUP))

    # ------------------------------------------------------------------ lookups
    def get(self, topic_id: str) -> Optional[Dict[str, Any]]:
        return self.topics.get(topic_id)

    def for_page(self, page: str) -> str:
        """The topic that explains the screen the user is on (context help)."""
        return self.context.get(page) or self.context.get((page or "").lower()) or "start-here"

    def related(self, topic_id: str) -> List[Dict[str, Any]]:
        t = self.get(topic_id) or {}
        return [self.topics[r] for r in t.get("related", []) if r in self.topics]

    def chapter_topics(self, chapter_id: str) -> List[Dict[str, Any]]:
        for c in self.chapters:
            if c["id"] == chapter_id:
                return [self.topics[t] for t in c["topics"] if t in self.topics]
        return []

    # ------------------------------------------------------------------ search
    def _prepare(self, t: Dict[str, Any]) -> Dict[str, Any]:
        title = tokens(t.get("title", ""))
        kw = tokens(" ".join(t.get("keywords", [])))
        heads = tokens(" ".join(t.get("headings", [])))
        body = tokens(t.get("text", ""))
        return {"title_x": set(title), "kw_x": set(kw),
                "title": {_canon(w) for w in title}, "kw": {_canon(w) for w in kw}, "heads": {_canon(w) for w in heads},
                "body": [_canon(w) for w in body], "all": set(title + kw + heads + body)}

    def _fix_typos(self, toks: List[str]) -> List[str]:
        fixed = []
        for w in toks:
            if w in self._vocab or len(w) < 4:
                fixed.append(w)
                continue
            near = difflib.get_close_matches(w, self._vocab, n=1, cutoff=0.78)
            fixed.append(near[0] if near else w)
        return fixed

    def search(self, query: str, limit: int = 8) -> List[Hit]:
        raw = tokens(query)
        if not raw:
            return []
        qtoks = self._fix_typos(raw)
        q = [_canon(w) for w in qtoks]
        qtext = " ".join((query or "").lower().split())
        scores: Dict[str, float] = {}
        why: Dict[str, str] = {}
        for tid, ix in self._index.items():
            s = 0.0
            for w in qtoks:                                   # the very word typed (not just a synonym of it) counts for more
                if w in ix["title_x"]:
                    s += 6
                if w in ix["kw_x"]:
                    s += 4
            for w in q:
                if w in ix["title"]:
                    s += 12
                if w in ix["kw"]:
                    s += 8
                if w in ix["heads"]:
                    s += 4
                s += min(ix["body"].count(w), 5) * 1.0
            if qtext and qtext in (self.topics[tid].get("title", "").lower()):
                s += 25
            if s:
                scores[tid] = s
        for intent in self.intents:                                    # "I want to ..." sentences written for exactly this purpose
            itoks = {_canon(w) for w in tokens(intent["phrase"])}
            if not itoks:
                continue
            overlap = len(itoks & set(q)) / max(len(itoks), 1)
            cover = len(itoks & set(q)) / max(len(set(q)), 1)
            if overlap >= 0.5 and cover >= 0.5:
                bonus = 30 * overlap * cover + (20 if qtext and qtext in intent["phrase"].lower() else 0)
                scores[intent["topic"]] = scores.get(intent["topic"], 0) + bonus
                why[intent["topic"]] = f"Matches \"{intent['phrase']}\""
        for f in self.faq:
            ftoks = {_canon(w) for w in tokens(f["q"])}
            if ftoks and len(ftoks & set(q)) / max(len(ftoks), 1) >= 0.5 and f.get("topic") in self.topics:
                scores[f["topic"]] = scores.get(f["topic"], 0) + 15
                why.setdefault(f["topic"], f"Answers: {f['q']}")
        ranked = sorted(scores.items(), key=lambda kv: kv[1], reverse=True)[:limit]
        hits = []
        for tid, sc in ranked:
            t = self.topics[tid]
            hits.append(Hit(t, round(sc, 1), self._snippet(t, qtoks), why.get(tid, ""), list(t.get("steps", []))[:6]))
        return hits

    def _snippet(self, t: Dict[str, Any], qtoks: List[str]) -> str:
        want = {_canon(w) for w in qtoks}
        best, best_n = t.get("summary", ""), 0
        for sent in re.split(r"(?<=[.!?])\s+|\n", t.get("text", "")):
            n = len(want & {_canon(w) for w in tokens(sent)})
            if n > best_n and 20 < len(sent) < 260:
                best, best_n = sent.strip(), n
        return best if best_n >= min(2, len(want)) else t.get("summary", "")

    def suggestions(self, limit: int = 6) -> List[str]:
        """Phrases to try: taken from the guide's own 'I want to...' list."""
        return [i["phrase"] for i in self.intents[:limit]]


import json
_DATA = json.loads("{\"title\": \"BillDesk — User Guide and Help\", \"chapters\": [{\"id\": \"start\", \"title\": \"1. Start here\", \"topics\": [\"start-here\", \"what-is\", \"signing-in\", \"main-window\", \"dashboard\"]}, {\"id\": \"billing\", \"title\": \"2. Billing at the counter\", \"topics\": [\"new-bill-overview\", \"bill-customer\", \"bill-date\", \"bill-items\", \"bill-duplicate\", \"bill-payment\", \"bill-park\", \"bill-print\"]}, {\"id\": \"history\", \"title\": \"3. Bill history and returns\", \"topics\": [\"history\", \"history-filters\", \"history-view\", \"history-void\", \"return-goods\", \"credit-notes\", \"cancel-return\"]}, {\"id\": \"orders\", \"title\": \"4. Customer orders\", \"topics\": [\"orders-list\", \"order-new\", \"order-dates\", \"smart-importer\", \"text-importer\", \"order-convert\", \"order-matrix\"]}, {\"id\": \"masters\", \"title\": \"5. Masters: items, customers, suppliers\", \"topics\": [\"items\", \"customers\", \"suppliers\", \"fixed-rates\"]}, {\"id\": \"stock\", \"title\": \"6. Stock and purchases\", \"topics\": [\"inventory\", \"stock-adjust\", \"waste\", \"crates\", \"procurement-bills\", \"vendor-bill\", \"purchase-orders\", \"purchase-return\"]}, {\"id\": \"accounts\", \"title\": \"7. Money and accounts\", \"topics\": [\"accounting-home\", \"receivables\", \"receipt\", \"statement\", \"payables\", \"supplier-payment\", \"banking\", \"journal\", \"statements\", \"integrity\"]}, {\"id\": \"reports\", \"title\": \"8. Reports\", \"topics\": [\"daybook\", \"item-wise\", \"customer-wise\", \"report-dates\", \"consolidated\"]}, {\"id\": \"admin\", \"title\": \"9. Administration\", \"topics\": [\"cash-drawer\", \"users\", \"company\", \"backup\", \"audit-logs\"]}, {\"id\": \"reference\", \"title\": \"10. Reference\", \"topics\": [\"shortcuts\", \"dates-everywhere\", \"field-rules\", \"messages\", \"faq\", \"glossary\", \"about-guide\"]}], \"intents\": [{\"phrase\": \"make a bill for a customer\", \"topic\": \"new-bill-overview\"}, {\"phrase\": \"sell to a walk-in customer for cash\", \"topic\": \"bill-customer\"}, {\"phrase\": \"change the bill date\", \"topic\": \"bill-date\"}, {\"phrase\": \"add items quickly with the keyboard\", \"topic\": \"bill-items\"}, {\"phrase\": \"put a bill on hold and serve someone else\", \"topic\": \"bill-park\"}, {\"phrase\": \"take payment or leave it on credit\", \"topic\": \"bill-payment\"}, {\"phrase\": \"print or save an invoice or delivery challan\", \"topic\": \"bill-print\"}, {\"phrase\": \"find an old bill\", \"topic\": \"history\"}, {\"phrase\": \"see who has not paid\", \"topic\": \"history-filters\"}, {\"phrase\": \"cancel a bill I made by mistake\", \"topic\": \"history-void\"}, {\"phrase\": \"take goods back from a customer\", \"topic\": \"return-goods\"}, {\"phrase\": \"cancel a return I entered by mistake\", \"topic\": \"cancel-return\"}, {\"phrase\": \"print a credit note\", \"topic\": \"credit-notes\"}, {\"phrase\": \"record a customer's order\", \"topic\": \"order-new\"}, {\"phrase\": \"read an order from a photo or WhatsApp image\", \"topic\": \"smart-importer\"}, {\"phrase\": \"turn an order into a bill\", \"topic\": \"order-convert\"}, {\"phrase\": \"know how much to buy for tomorrow\", \"topic\": \"order-matrix\"}, {\"phrase\": \"add a customer, item or supplier\", \"topic\": \"customers\"}, {\"phrase\": \"give a customer a fixed rate\", \"topic\": \"fixed-rates\"}, {\"phrase\": \"correct stock after counting\", \"topic\": \"stock-adjust\"}, {\"phrase\": \"write off rotten produce\", \"topic\": \"waste\"}, {\"phrase\": \"enter a supplier's bill\", \"topic\": \"vendor-bill\"}, {\"phrase\": \"send goods back to a supplier\", \"topic\": \"purchase-return\"}, {\"phrase\": \"record money received from a customer\", \"topic\": \"receipt\"}, {\"phrase\": \"see what a customer owes, bill by bill\", \"topic\": \"statement\"}, {\"phrase\": \"pay a supplier\", \"topic\": \"supplier-payment\"}, {\"phrase\": \"see today's money in and out\", \"topic\": \"daybook\"}, {\"phrase\": \"see which items sell best\", \"topic\": \"item-wise\"}, {\"phrase\": \"count the cash drawer\", \"topic\": \"cash-drawer\"}, {\"phrase\": \"add a user or change who can do what\", \"topic\": \"users\"}, {\"phrase\": \"change the company name, address or logo\", \"topic\": \"company\"}, {\"phrase\": \"back up my data\", \"topic\": \"backup\"}, {\"phrase\": \"check that the books add up\", \"topic\": \"integrity\"}, {\"phrase\": \"understand an error message\", \"topic\": \"messages\"}], \"faq\": [{\"q\": \"\\\"Choose a customer from the customer master (F5)\\\"\", \"a\": \"an order needs a real customer press F5, choose one; add the customer first if missing\", \"topic\": \"messages\"}, {\"q\": \"\\\"The bill date cannot be in the future\\\"\", \"a\": \"the date box holds a later day pick today or an earlier day\", \"topic\": \"messages\"}, {\"q\": \"\\\"… is not a valid date. Use DD/MM/YYYY\\\"\", \"a\": \"text in a date box can't be read pick from the calendar\", \"topic\": \"messages\"}, {\"q\": \"\\\"The To date cannot be before …\\\"\", \"a\": \"To is earlier than From change one of them\", \"topic\": \"messages\"}, {\"q\": \"\\\"Amount received (…) is more than the bill total\\\"\", \"a\": \"more entered than the bill type the bill amount or less\", \"topic\": \"messages\"}, {\"q\": \"\\\"Item … not found\\\" / \\\"No item matches\\\"\", \"a\": \"wrong code check the code in the Item Master\", \"topic\": \"messages\"}, {\"q\": \"\\\"… must be a number\\\"\", \"a\": \"letters or symbols in a number box type digits only (decimals allowed)\", \"topic\": \"messages\"}, {\"q\": \"\\\"An item / customer named … already exists\\\"\", \"a\": \"the name is taken use another name, or edit the existing one\", \"topic\": \"messages\"}, {\"q\": \"\\\"Vendor invoice … is already recorded as PUR-…\\\"\", \"a\": \"the same supplier invoice was entered before do not enter it again\", \"topic\": \"messages\"}, {\"q\": \"\\\"Payment amount … exceeds purchase balance due\\\"\", \"a\": \"paying a supplier more than owed pay the balance due or less\", \"topic\": \"messages\"}, {\"q\": \"\\\"Waste quantity … exceeds stock on hand\\\"\", \"a\": \"more than the tracked stock check the quantity\", \"topic\": \"messages\"}, {\"q\": \"\\\"Invoice … has goods returned against it, so it cannot be voided\\\"\", \"a\": \"a return exists cancel the return first, then void\", \"topic\": \"cancel-return\"}, {\"q\": \"\\\"only n can still be returned\\\"\", \"a\": \"more than billed minus earlier returns reduce the return quantity\", \"topic\": \"messages\"}, {\"q\": \"\\\"Return … was made by the older BillDesk and cannot be cancelled here\\\"\", \"a\": \"old returns lack detail leave it; it is already reflected in the balance\", \"topic\": \"messages\"}, {\"q\": \"\\\"The entry does not balance\\\"\", \"a\": \"journal debit ≠ credit make the two amounts equal\", \"topic\": \"messages\"}, {\"q\": \"\\\"You already have an open cashier session\\\"\", \"a\": \"a drawer is open close it first\", \"topic\": \"messages\"}, {\"q\": \"\\\"Access Denied\\\"\", \"a\": \"your role does not include that screen ask the administrator\", \"topic\": \"messages\"}, {\"q\": \"\\\"Tesseract OCR engine is not installed\\\"\", \"a\": \"the image reader is missing use Paste text instead, or ask IT to reinstall\", \"topic\": \"messages\"}, {\"q\": \"\\\"The text in this image could not be read reliably\\\"\", \"a\": \"handwriting or a blurry photo take a straighter, brighter photo or type the lines\", \"topic\": \"messages\"}, {\"q\": \"Integrity check shows Warning\", \"a\": \"an older-data difference open the line; ask your accountant if unsure\", \"topic\": \"messages\"}, {\"q\": \"Cannot start: database error\", \"a\": \"the database did not start restart the computer, then BillDesk; call IT if it repeats\", \"topic\": \"messages\"}, {\"q\": \"Can I change the date of a bill after saving?\", \"a\": \"No. Void the bill and make it again with the right date.\", \"topic\": \"history-void\"}, {\"q\": \"How do I sell to someone who is not a customer?\", \"a\": \"Leave Cash as the customer and take the full payment.\", \"topic\": \"bill-customer\"}, {\"q\": \"Why did the bill number have a different date?\", \"a\": \"The invoice number follows the bill date you chose.\", \"topic\": \"bill-date\"}, {\"q\": \"A customer paid half. What now?\", \"a\": \"Enter the amount received; the rest stays on their account.\", \"topic\": \"bill-payment\"}, {\"q\": \"A customer paid later in cash. How do I record it?\", \"a\": \"Use Record Customer Receipt.\", \"topic\": \"receipt\"}, {\"q\": \"How do I see what one customer owes, bill by bill?\", \"a\": \"Open the customer's Statement.\", \"topic\": \"statement\"}, {\"q\": \"What is \\\"Cr\\\" next to a balance?\", \"a\": \"You hold the customer's money (advance or return credit).\", \"topic\": \"customers\"}, {\"q\": \"What is the difference between void and return?\", \"a\": \"Void cancels the whole bill; return takes back some goods and keeps the bill.\", \"topic\": \"return-goods\"}, {\"q\": \"I returned goods by mistake.\", \"a\": \"Cancel the return.\", \"topic\": \"cancel-return\"}, {\"q\": \"Does BillDesk handle GST?\", \"a\": \"No: fruit and vegetables are exempt, so no GST is calculated.\", \"topic\": \"what-is\"}, {\"q\": \"Can two people bill at once?\", \"a\": \"Yes on different computers using the same database; park a bill to serve two customers at one counter.\", \"topic\": \"bill-park\"}, {\"q\": \"How often should I back up?\", \"a\": \"BillDesk backs up daily; also copy the backup folder to another drive weekly.\", \"topic\": \"backup\"}, {\"q\": \"Can it read my customer's handwritten list?\", \"a\": \"Not reliably. Printed or typed lists work; otherwise type or paste.\", \"topic\": \"smart-importer\"}, {\"q\": \"Stock shows 0 but I can still sell. Is that wrong?\", \"a\": \"No. Stock only tracks what you entered through purchases or adjustments.\", \"topic\": \"inventory\"}, {\"q\": \"I forgot my password.\", \"a\": \"Ask the administrator; they reset it.\", \"topic\": \"signing-in\"}, {\"q\": \"Where do I change my shop name and address on invoices?\", \"a\": \"Settings → Company Settings.\", \"topic\": \"company\"}], \"context\": {\"Dashboard\": \"dashboard\", \"New Bill\": \"new-bill-overview\", \"Bill History\": \"history\", \"Orders\": \"orders-list\", \"New Order\": \"order-new\", \"Order Matrix\": \"order-matrix\", \"Item Master\": \"items\", \"Customer Master\": \"customers\", \"Supplier Master\": \"suppliers\", \"Fixed Rates\": \"fixed-rates\", \"Inventory\": \"inventory\", \"Waste Management\": \"inventory\", \"Procurement\": \"procurement-bills\", \"Finance\": \"accounting-home\", \"Accounts Receivables\": \"receivables\", \"Accounts Payables\": \"payables\", \"BRS\": \"banking\", \"Trial Balance\": \"statements\", \"Profit & Loss\": \"statements\", \"Balance Sheet\": \"statements\", \"Integrity Check\": \"integrity\", \"Daybook\": \"daybook\", \"Reports\": \"daybook\", \"Item-wise Sales\": \"item-wise\", \"Customer-wise Sales\": \"customer-wise\", \"Consolidated Billing\": \"consolidated\", \"Handover & Settlement\": \"cash-drawer\", \"User Management\": \"users\", \"Company Settings\": \"company\", \"DB Connection\": \"backup\", \"System Audit Logs\": \"audit-logs\"}, \"topics\": {\"start-here\": {\"id\": \"start-here\", \"title\": \"How to use this guide\", \"number\": \"1.1\", \"chapter\": \"start\", \"chapter_title\": \"Start here\", \"screen\": \"Help\", \"keywords\": [\"help\", \"guide\", \"search\", \"find\", \"how to use\", \"index\", \"contents\", \"F9\", \"ask\"], \"related\": [\"shortcuts\", \"faq\", \"messages\"], \"summary\": \"This guide is written so you can ask a question instead of reading it from page one.\", \"steps\": [\"In BillDesk press **F9** (on any screen it opens the help for *that* screen) or choose **Help → Help & User Guide**.\", \"Type what you want in your own words, for example *how do I return goods*, *cancel a bill*, *customer owes money*, *backup*. Press **Enter**.\", \"Click the best result. Each result shows the first steps right away; the topic page has the pictures and details.\", \"Use **◀ Back** to return to your results, and **Contents** to browse by chapter.\"], \"headings\": [\"make a bill for a customer\", \"sell to a walk-in customer for cash\", \"change the bill date\", \"add items quickly with the keyboard\", \"put a bill on hold and serve someone else\", \"take payment or leave it on credit\", \"print or save an invoice or delivery challan\", \"find an old bill\", \"see who has not paid\", \"cancel a bill I made by mistake\", \"take goods back from a customer\", \"cancel a return I entered by mistake\", \"print a credit note\", \"record a customer's order\", \"read an order from a photo or WhatsApp image\", \"turn an order into a bill\", \"know how much to buy for tomorrow\", \"add a customer, item or supplier\", \"give a customer a fixed rate\", \"correct stock after counting\", \"write off rotten produce\", \"enter a supplier's bill\", \"send goods back to a supplier\", \"record money received from a customer\", \"see what a customer owes, bill by bill\", \"pay a supplier\", \"see today's money in and out\", \"see which items sell best\", \"count the cash drawer\", \"add a user or change who can do what\", \"change the company name, address or logo\", \"back up my data\", \"check that the books add up\", \"understand an error message\"], \"text\": \"This guide is written so you can ask a question instead of reading it from page one.\\nIn BillDesk press F9 (on any screen it opens the help for that screen) or choose Help → Help & User Guide.\\nType what you want in your own words, for example how do I return goods, cancel a bill, customer owes money, backup. Press Enter.\\nClick the best result. Each result shows the first steps right away; the topic page has the pictures and details.\\nUse ◀ Back to return to your results, and Contents to browse by chapter.\\nF9 on the New Bill screen opens the help for that screen\\nTyping a question lists the best topics first, with the first steps of each so you may not need to open it:\\nSearching the help\\nOn paper, in Word, in the PDF or in the notebook, use Ctrl+F and the same words. The notebook edition also has a ask(\\\"...\\\") search (see the last cell).\\nFind it by what you want to do\\nI want to… | Go to\\nmake a bill for a customer | Making a bill\\nsell to a walk-in customer for cash | Choosing the customer\\nchange the bill date | The bill date and the calendar\\nadd items quickly with the keyboard | Entering items\\nput a bill on hold and serve someone else | Park a bill and recall it\\ntake payment or leave it on credit | Taking payment\\nprint or save an invoice or delivery challan | Print and PDF\\nfind an old bill | Bill History\\nsee who has not paid | Filtering bills\\ncancel a bill I made by mistake | Void a bill\\ntake goods back from a customer | Return goods (credit note)\\ncancel a return I entered by mistake | Cancel a return\\nprint a credit note | Credit notes\\nrecord a customer's order | New customer order\\nread an order from a photo or WhatsApp image | Smart Importer\\nturn an order into a bill | Convert an order\\nknow how much to buy for tomorrow | Order matrix\\nadd a customer, item or supplier | Customers, Items, Suppliers\\ngive a customer a fixed rate | Fixed contract rates\\ncorrect stock after counting | Stock adjustment\\nwrite off rotten produce | Waste and spoilage\\nenter a supplier's bill | Vendor bills\\nsend goods back to a supplier | Return to supplier (debit note)\\nrecord money received from a customer | Customer receipts\\nsee what a customer owes, bill by bill | Customer statement\\npay a supplier | Supplier payments\\nsee today's money in and out | Daybook\\nsee which items sell best | Item-wise sales\\ncount the cash drawer | Cash drawer\\nadd a user or change who can do what | Users and roles\\nchange the company name, address or logo | Company configuration\\nback up my data | Backup and database\\ncheck that the books add up | Integrity check\\nunderstand an error message | Messages and what to do\\nStill stuck? See Frequently asked questions and Messages and what to do.\"}, \"what-is\": {\"id\": \"what-is\", \"title\": \"What BillDesk does\", \"number\": \"1.2\", \"chapter\": \"start\", \"chapter_title\": \"Start here\", \"screen\": \"All\", \"keywords\": [\"overview\", \"about\", \"what is billdesk\", \"features\", \"accounts\", \"automatic\", \"GST\"], \"related\": [\"dashboard\", \"signing-in\"], \"summary\": \"BillDesk is a desktop program for the billing counter and the back office of a wholesale fruit and vegetable business. One program, one database, one set of numbers.\", \"steps\": [], \"headings\": [\"Make a bill\", \"Receive payment\", \"Enter a supplier bill\", \"Return goods\", \"Void a bill\", \"Take a customer order\"], \"text\": \"BillDesk is a desktop program for the billing counter and the back office of a wholesale fruit and vegetable business. One program, one database, one set of numbers.\\nYou do this… | …and BillDesk automatically\\nMake a bill | reduces stock, adds to the customer's balance, records the sale and its cost in the accounts\\nReceive payment | reduces the customer's balance, records cash or bank in the accounts, closes the bill\\nEnter a supplier bill | adds what you owe the supplier (and TDS if it applies)\\nReturn goods | credits the customer, puts good stock back, reverses the sale in the accounts\\nVoid a bill | puts the stock back, reverses the balance and the accounts\\nTake a customer order | lists it in the order matrix so you know how much to buy, and turns it into a bill with one click\\nGST is not used (fresh fruit and vegetables are exempt), so there is no tax set-up. A GSTIN can still be stored for a customer or supplier if you want it on file.\\nWho sees what. Menus depend on your role. Cashiers normally see Billing, Orders and Bill History; managers also see Masters, Reports and Accounts; administrators see everything including Settings. If a menu is missing, your role does not include it: ask the administrator (Users and roles).\"}, \"signing-in\": {\"id\": \"signing-in\", \"title\": \"Signing in\", \"number\": \"1.3\", \"chapter\": \"start\", \"chapter_title\": \"Start here\", \"screen\": \"Login\", \"keywords\": [\"login\", \"sign in\", \"username\", \"password\", \"forgot password\", \"admin\", \"first time\", \"log out\", \"logout\", \"F12\"], \"related\": [\"users\", \"main-window\", \"faq\"], \"summary\": \"Double-click BillDesk on the desktop or in the Start menu. The first start takes a few seconds because BillDesk starts its own database.\", \"steps\": [\"Double-click **BillDesk** on the desktop or in the Start menu. The first start takes a few seconds because BillDesk starts its own database.\", \"Type your **username** and **password**.\", \"Press **Enter** or click **Login**.\"], \"headings\": [], \"text\": \"Double-click BillDesk on the desktop or in the Start menu. The first start takes a few seconds because BillDesk starts its own database.\\nType your username and password.\\nPress Enter or click Login.\\nSign-in screen\\nFirst time ever: the installer creates one administrator named admin with a random password shown once. Write it down, then create the other users in Users and roles.\\nForgot the administrator password? Ask your IT person to run python scripts\\\\reset_password.py --user admin.\\nOnly one BillDesk window can be open on a computer; if you start it twice, the second copy tells you.\\nTo sign out press F12 (you are asked to confirm).\"}, \"main-window\": {\"id\": \"main-window\", \"title\": \"Finding your way around\", \"number\": \"1.4\", \"chapter\": \"start\", \"chapter_title\": \"Start here\", \"screen\": \"Main window\", \"keywords\": [\"menu\", \"menu bar\", \"navigation\", \"layout\", \"shortcut strip\", \"keys\", \"screens\", \"where is\"], \"related\": [\"shortcuts\", \"dashboard\", \"start-here\"], \"summary\": \"Press F9 on any screen for help about that screen. All shortcut keys are listed in Keyboard shortcuts.\", \"steps\": [], \"headings\": [\"File\", \"Masters\", \"Reports\", \"Accounts\", \"Settings\", \"Help\"], \"text\": \"Menu bar\\nTop: your company name and the menu bar.\\nMiddle: the screen you are working on.\\nBottom strip: the keys that work on this screen, who is signed in and the date and time.\\nMenu | Contains\\nFile | New Bill (F2), New Customer Order, Ordering System (Ctrl+O), Purchase system (Ctrl+P), Logout (F12), Exit\\nMasters | Item Master (F10), Customer Master (F4), Supplier Master (F11), Inventory & Stock, Waste Management\\nReports | Bills History (F6), Daybook, Item-wise Sales, Customer-wise Sales, Bills Consolidated Report, Order Consolidation, Fixed Rates Report, Profit & Loss, Balance Sheet, Dashboard (F1)\\nAccounts | Accounting Dashboard (Ctrl+D), Trial Balance, Profit & Loss, Balance Sheet, BRS, Handover & Settlement, Integrity Check, Accounts Receivables, Accounts Payables\\nSettings | User Management, Company Settings, Database Settings, System Audit Logs\\nHelp | Help & User Guide (F9), Keyboard shortcuts, About\\nPress F9 on any screen for help about that screen. All shortcut keys are listed in Keyboard shortcuts.\"}, \"dashboard\": {\"id\": \"dashboard\", \"title\": \"The dashboard\", \"number\": \"1.5\", \"chapter\": \"start\", \"chapter_title\": \"Start here\", \"screen\": \"Dashboard\", \"keywords\": [\"dashboard\", \"home\", \"sales today\", \"analytics\", \"cards\", \"quick actions\", \"at a glance\", \"receivables\", \"overdue\", \"collected today\", \"F1\"], \"related\": [\"new-bill-overview\", \"receivables\", \"daybook\"], \"summary\": \"The dashboard is the first screen after you sign in. Press F1 from anywhere to come back to it (the figures are re-read each time).\", \"steps\": [], \"headings\": [\"To collect (receivables)\", \"Overdue over 30 days\", \"Collected today\", \"Orders waiting\", \"Sales – last 14 days\", \"Customers who owe the most\"], \"text\": \"The dashboard is the first screen after you sign in. Press F1 from anywhere to come back to it (the figures are re-read each time).\\nDashboard\\nSales, Order and Purchase Analytics: six small cards each: today, yesterday, this week, last week, this month, last month. Every card shows the amount and how many bills or orders it covers.\\nQuick Actions: big buttons for New Bill, Items, Customers, Suppliers and Existing Bills.\\nAt a glance (scroll down):\\nAt a glance\\nCard | Meaning\\nTo collect (receivables) | everything customers owe you now (customers who have paid in advance are shown separately as \\\"advances held\\\", not subtracted)\\nOverdue over 30 days | the part of that which is older than 30 days\\nCollected today | receipts recorded today\\nOrders waiting | customer orders not yet billed\\nSales – last 14 days | one bar per day\\nCustomers who owe the most | the top four balances\"}, \"new-bill-overview\": {\"id\": \"new-bill-overview\", \"title\": \"Making a bill\", \"number\": \"2.1\", \"chapter\": \"billing\", \"chapter_title\": \"Billing at the counter\", \"screen\": \"New Bill\", \"keywords\": [\"new bill\", \"invoice\", \"billing\", \"sale\", \"make a bill\", \"counter\", \"F2\", \"grid\", \"cash bill\", \"credit bill\", \"how to bill\"], \"related\": [\"bill-customer\", \"bill-date\", \"bill-items\", \"bill-payment\"], \"summary\": \"Open File → New Bill or press F2.\", \"steps\": [\"Press **F5** and choose the customer (or leave **Cash** for a walk-in).\", \"The cursor moves to the **Date**; accept today or pick another day from the calendar.\", \"Type each item's **code** and press **Enter**; type the **quantity**, **Enter**; the unit and the rate follow.\", \"Press **F2** to take payment (or **F3** to save **and** print).\", \"Check the amount received and press **Post Payment**.\"], \"headings\": [\"Billing Company\", \"Customer (F5)\", \"Date\", \"The grid\", \"Line under the grid\", \"Customer (Delivery) / Bill To\", \"Internal Notes\", \"Total\", \"Park (F6) / Parked (F7)\"], \"text\": \"Open File → New Bill or press F2.\\nPress F5 and choose the customer (or leave Cash for a walk-in).\\nThe cursor moves to the Date; accept today or pick another day from the calendar.\\nType each item's code and press Enter; type the quantity, Enter; the unit and the rate follow.\\nPress F2 to take payment (or F3 to save and print).\\nCheck the amount received and press Post Payment.\\nNew Bill, empty\\nPart of the screen | What it is\\nBilling Company | which of your companies the invoice is issued under\\nCustomer (F5) | who the bill is for (choosing the customer)\\nDate | the bill date, with a drop-down calendar (dates)\\nThe grid | one row per item: Code, Item Description, Qty, Unit, Rate, Amount, and a red ✕ to clear the row\\nLine under the grid | stock and rate of the item you just entered (the hint line)\\nCustomer (Delivery) / Bill To | where the goods go and who is invoiced\\nInternal Notes | for your own use, not printed\\nTotal | the bill total\\nPark (F6) / Parked (F7) | put a bill on hold, or bring one back\\nA bill with six items\"}, \"bill-customer\": {\"id\": \"bill-customer\", \"title\": \"Choosing the customer\", \"number\": \"2.2\", \"chapter\": \"billing\", \"chapter_title\": \"Billing at the counter\", \"screen\": \"New Bill\", \"keywords\": [\"customer\", \"F5\", \"search customer\", \"cash customer\", \"walk-in\", \"cash sale\", \"pick customer\", \"select customer\"], \"related\": [\"new-bill-overview\", \"bill-date\", \"customers\"], \"summary\": \"Press F5 (or click the customer box).\", \"steps\": [\"Type part of the customer's name or phone number: the list narrows as you type.\", \"Press **↓** or **↑** to move, **Enter** to choose. (Esc closes the list; a click also chooses.)\", \"Choose **Cash Customer** for a walk-in sale.\"], \"headings\": [], \"text\": \"Press F5 (or click the customer box).\\nType part of the customer's name or phone number: the list narrows as you type.\\nPress ↓ or ↑ to move, Enter to choose. (Esc closes the list; a click also chooses.)\\nChoose Cash Customer for a walk-in sale.\\nCustomer search\\nThe customer's address, their \\\"bill to\\\" name and their company are filled in for you, and any fixed contract rates for that customer are used.\\nThe same dialog is used on New customer order, so it works the same everywhere. Orders need a real customer (no Cash option there).\\nA customer who is not in the list must be added first in the Customer Master.\"}, \"bill-date\": {\"id\": \"bill-date\", \"title\": \"The bill date and the calendar\", \"number\": \"2.3\", \"chapter\": \"billing\", \"chapter_title\": \"Billing at the counter\", \"screen\": \"New Bill\", \"keywords\": [\"date\", \"calendar\", \"bill date\", \"back date\", \"previous day\", \"date picker\", \"wrong date\", \"future date\"], \"related\": [\"dates-everywhere\", \"bill-items\"], \"summary\": \"Right after you choose the customer the cursor goes to Date and a calendar drops down under it.\", \"steps\": [\"Click a day, or use the arrow keys and press **Enter**. **Today** jumps to today.\", \"Or type a date (`13/10/2026`, `13-10-26`) and press **Enter**.\", \"The cursor moves on to the first empty row of the grid.\"], \"headings\": [], \"text\": \"Right after you choose the customer the cursor goes to Date and a calendar drops down under it.\\nClick a day, or use the arrow keys and press Enter. Today jumps to today.\\nOr type a date (13/10/2026, 13-10-26) and press Enter.\\nThe cursor moves on to the first empty row of the grid.\\nCalendar under the date box\\nThe date you choose is the date of the bill and its invoice number follows that day (20261005-0003 for 5 October).\\nA bill cannot be dated in the future.\\nMore about date boxes everywhere in the program: Dates and the calendar.\"}, \"bill-items\": {\"id\": \"bill-items\", \"title\": \"Entering items\", \"number\": \"2.4\", \"chapter\": \"billing\", \"chapter_title\": \"Billing at the counter\", \"screen\": \"New Bill\", \"keywords\": [\"items\", \"code\", \"quantity\", \"qty\", \"unit\", \"rate\", \"enter key\", \"keyboard\", \"hint line\", \"stock\", \"fixed rate\", \"wrong item\", \"choose item\", \"which item\"], \"related\": [\"bill-duplicate\", \"bill-payment\", \"fixed-rates\"], \"summary\": \"Work row by row with the keyboard: Code → Qty → Unit → Rate → next row, pressing Enter each time.\", \"steps\": [\"Type the item **code** (for example `111`) or a part of its **name** (`tom`), press **Enter**. The description, unit and rate appear and the cursor goes to **Qty** (1 is filled in).\", \"Type the **quantity**, press **Enter**. The cursor passes through **Unit** to **Rate**.\", \"Change the **rate** if you need to; press **Enter** to go to the next row.\"], \"headings\": [], \"text\": \"Work row by row with the keyboard: Code → Qty → Unit → Rate → next row, pressing Enter each time.\\nType the item code (for example 111) or a part of its name (tom), press Enter. The description, unit and rate appear and the cursor goes to Qty (1 is filled in).\\nType the quantity, press Enter. The cursor passes through Unit to Rate.\\nChange the rate if you need to; press Enter to go to the next row.\\nThe hint line\\nThe hint line under the grid tells you the item, its stock (\\\"No stock recorded\\\" if stock is not tracked for it) and the rate and where it came from: fixed rate for <customer> or item rate.\\nSeveral items fit what you typed? A list opens: ↑ / ↓ to move, Enter to choose, Esc to cancel. BillDesk never picks the first match for you.\\nChoose between items\\nAn item typed in a lower row moves up to the first empty row, so there are never gaps.\\nThe red ✕ clears a row and the rows below move up.\\nYou can sell more than the stock shows; BillDesk warns but does not stop you (your administrator can change this).\\nQuantity and rate must be numbers; a bill cannot have a zero quantity or a negative rate.\"}, \"bill-duplicate\": {\"id\": \"bill-duplicate\", \"title\": \"The same item twice\", \"number\": \"2.5\", \"chapter\": \"billing\", \"chapter_title\": \"Billing at the counter\", \"screen\": \"New Bill\", \"keywords\": [\"duplicate\", \"same item twice\", \"add quantity\", \"repeated item\"], \"related\": [\"bill-items\"], \"summary\": \"If you type an item that is already on the bill, BillDesk asks what to do.\", \"steps\": [], \"headings\": [], \"text\": \"If you type an item that is already on the bill, BillDesk asks what to do.\\nDuplicate item\\nAdd the QTY (Enter) adds the new quantity to the existing row.\\nIgnore (Delete Duplicate) removes the new row.\"}, \"bill-payment\": {\"id\": \"bill-payment\", \"title\": \"Taking payment\", \"number\": \"2.6\", \"chapter\": \"billing\", \"chapter_title\": \"Billing at the counter\", \"screen\": \"New Bill\", \"keywords\": [\"payment\", \"pay\", \"post payment\", \"F2\", \"F3\", \"amount received\", \"cash\", \"UPI\", \"cheque\", \"credit\", \"due\", \"balance\", \"outstanding\", \"save bill\", \"reference\", \"UTR\"], \"related\": [\"bill-print\", \"receipt\", \"statement\"], \"summary\": \"Press F2 (save) or F3 (save and print).\", \"steps\": [\"Press **F2** (save) or **F3** (save and print).\", \"The amount received is filled in with the whole total. Change it if the customer pays part.\", \"Choose the **payment method** and, for UPI or a cheque, the **reference**.\", \"Press **Enter** or **Post Payment**.\"], \"headings\": [\"Outstanding / Unpaid bills\", \"Amount Received\", \"Payment Date\", \"Payment Method\", \"Reference / UTR\"], \"text\": \"Press F2 (save) or F3 (save and print).\\nThe amount received is filled in with the whole total. Change it if the customer pays part.\\nChoose the payment method and, for UPI or a cheque, the reference.\\nPress Enter or Post Payment.\\nPayment dialog\\nBox | Meaning\\nOutstanding / Unpaid bills | what this customer already owes before this bill\\nAmount Received | what the customer pays now; anything not paid stays on their account\\nPayment Date | shown for information: a counter payment is recorded on the bill's date\\nPayment Method | Cash, UPI, Cheque, NEFT/RTGS, or Credit/Due (nothing paid now)\\nReference / UTR | UPI or cheque number (optional)\\nYou cannot receive more than the bill total.\\nAfter posting, the sheet is cleared for the next bill. With F3 the invoice opens for printing.\\nLater payments are recorded as a customer receipt.\"}, \"bill-park\": {\"id\": \"bill-park\", \"title\": \"Park a bill and recall it\", \"number\": \"2.7\", \"chapter\": \"billing\", \"chapter_title\": \"Billing at the counter\", \"screen\": \"New Bill\", \"keywords\": [\"park\", \"hold\", \"recall\", \"parked bills\", \"F6\", \"F7\", \"pause\", \"wait\", \"serve another customer\"], \"related\": [\"new-bill-overview\"], \"summary\": \"Serving two customers at once? Put one aside.\", \"steps\": [\"Press **F6** (or **Park**) with the items entered. The sheet clears.\", \"Serve the next customer.\", \"Press **F7** (or **Parked (n)**), click the parked bill to bring it back. If another bill is on the sheet you are asked before it is replaced.\"], \"headings\": [], \"text\": \"Serving two customers at once? Put one aside.\\nPress F6 (or Park) with the items entered. The sheet clears.\\nServe the next customer.\\nPress F7 (or Parked (n)), click the parked bill to bring it back. If another bill is on the sheet you are asked before it is replaced.\\nParked bills\\nParked bills are kept on disk: they survive closing BillDesk or a power cut.\"}, \"bill-print\": {\"id\": \"bill-print\", \"title\": \"Print and PDF\", \"number\": \"2.8\", \"chapter\": \"billing\", \"chapter_title\": \"Billing at the counter\", \"screen\": \"New Bill, Bill History\", \"keywords\": [\"print\", \"invoice\", \"delivery challan\", \"DC\", \"pdf\", \"preview\", \"save pdf\", \"printer\", \"copy\"], \"related\": [\"bill-payment\", \"history-view\"], \"summary\": \"After F3, or from Bill History, the invoice opens in the print preview.\", \"steps\": [], \"headings\": [], \"text\": \"After F3, or from Bill History, the invoice opens in the print preview.\\nPrint preview\\n− / + / Fit change the zoom; Print sends it to the printer; Save PDF saves a file; Close leaves.\\nThe invoice shows the company, customer, items with rates and amounts, the total in words and your terms:\\nInvoice\\nThe delivery challan (DC) is the same bill without prices, for the driver and the customer's receiving staff:\\nDelivery challan\"}, \"history\": {\"id\": \"history\", \"title\": \"Bill History\", \"number\": \"3.1\", \"chapter\": \"history\", \"chapter_title\": \"Bill history and returns\", \"screen\": \"Bill History\", \"keywords\": [\"bill history\", \"old bills\", \"find bill\", \"search bill\", \"invoices\", \"F6\", \"list of bills\", \"today's bills\"], \"related\": [\"history-filters\", \"history-view\", \"history-void\", \"return-goods\"], \"summary\": \"Open Reports → Bills History or press F6.\", \"steps\": [], \"headings\": [\"Preview Invoice / Preview DC\", \"Save Invoice / Save DC\", \"View Bill\", \"Void Bill\", \"Return Goods\", \"Credit Notes\", \"Refresh\"], \"text\": \"Open Reports → Bills History or press F6.\\nBill History\\nThe three coloured cards show the number of bills, total sales and today's bills.\\nThe table lists the newest bills first: invoice number, date, company, customer, number of items, amount and status.\\nSelect a bill, then use the buttons at the bottom (or right-click it):\\nButton | What it does\\nPreview Invoice / Preview DC | opens the invoice or the delivery challan\\nSave Invoice / Save DC | saves the PDF\\nView Bill | shows the bill's details\\nVoid Bill | cancels the bill (void)\\nReturn Goods | takes goods back (return)\\nCredit Notes | lists and prints the returns on this bill (credit notes)\\nRefresh | reloads the list\\nPer page and Previous / Next move through long lists.\"}, \"history-filters\": {\"id\": \"history-filters\", \"title\": \"Searching and filtering bills\", \"number\": \"3.2\", \"chapter\": \"history\", \"chapter_title\": \"Bill history and returns\", \"screen\": \"Bill History\", \"keywords\": [\"filter\", \"search\", \"status\", \"paid\", \"unpaid\", \"partial\", \"void\", \"legacy\", \"date filter\", \"find unpaid bills\", \"who has not paid\", \"outstanding bills\"], \"related\": [\"history\", \"statement\", \"receivables\"], \"summary\": \"To see what a customer owes bill by bill use the customer statement.\", \"steps\": [], \"headings\": [\"paid\", \"unpaid / partial\", \"void\", \"legacy\"], \"text\": \"Search (F1): type part of an invoice number or customer name.\\nStatus list: All, Unpaid, Partial, Paid, Void or Legacy.\\nOnly unpaid bills\\nInvoice date: click the box and pick a day from the calendar; Reset Date clears it. You can also type part of a date (09/2026) to see a month.\\nDate filter\\nStatus | Meaning\\npaid | fully paid\\nunpaid / partial | nothing / part paid; the rest is due\\nvoid | cancelled\\nlegacy | bills made by the older program: it did not track payment per bill; their money is in the customer's balance\\nTo see what a customer owes bill by bill use the customer statement.\"}, \"history-view\": {\"id\": \"history-view\", \"title\": \"Viewing a bill\", \"number\": \"3.3\", \"chapter\": \"history\", \"chapter_title\": \"Bill history and returns\", \"screen\": \"Bill History\", \"keywords\": [\"view bill\", \"bill details\", \"items on bill\", \"double click\"], \"related\": [\"history\", \"bill-print\"], \"summary\": \"Double-click a bill (or select it and press View Bill).\", \"steps\": [], \"headings\": [], \"text\": \"Double-click a bill (or select it and press View Bill).\\nBill details\\nIt shows the customer, date, status, the items with quantity, rate and amount and the grand total, with buttons to preview or save the invoice and the challan.\"}, \"history-void\": {\"id\": \"history-void\", \"title\": \"Void a bill\", \"number\": \"3.4\", \"chapter\": \"history\", \"chapter_title\": \"Bill history and returns\", \"screen\": \"Bill History\", \"keywords\": [\"void\", \"cancel bill\", \"delete bill\", \"wrong bill\", \"mistake\", \"reverse bill\", \"undo bill\"], \"related\": [\"return-goods\", \"history\"], \"summary\": \"A bill made by mistake is voided, not deleted, so the record and the invoice number stay.\", \"steps\": [\"Open [Bill History](#history) and select the bill.\", \"Press **Void Bill**.\", \"Read what will happen and confirm.\"], \"headings\": [], \"text\": \"A bill made by mistake is voided, not deleted, so the record and the invoice number stay.\\nOpen Bill History and select the bill.\\nPress Void Bill.\\nRead what will happen and confirm.\\nVoiding: reverses the customer's balance, puts the stock back, reverses the crates and the accounts, and marks the bill void. Payments that had been allocated to it become the customer's advance (credit).\\nA bill that has goods returned against it cannot be voided: cancel the return first.\\nA voided bill cannot be voided twice.\\nTo take back only some goods, use Return goods, not void.\"}, \"return-goods\": {\"id\": \"return-goods\", \"title\": \"Return goods (credit note)\", \"number\": \"3.5\", \"chapter\": \"history\", \"chapter_title\": \"Bill history and returns\", \"screen\": \"Bill History\", \"keywords\": [\"return\", \"return goods\", \"credit note\", \"goods back\", \"customer returned\", \"rotten\", \"damaged\", \"refund\", \"take back\", \"spoiled\", \"sales return\"], \"related\": [\"credit-notes\", \"cancel-return\", \"history-void\", \"statement\"], \"summary\": \"A customer brings goods back.\", \"steps\": [\"In [Bill History](#history) select the invoice they were bought on.\", \"Press **Return Goods**.\", \"Type the **Return qty** for each item. Tick **Spoiled?** if the goods are rotten and cannot be sold again.\", \"Write a **reason** (optional) and, for a walk-in customer, choose how you refund (Cash, UPI, Bank).\", \"Press **Save return**. You are offered the **credit note** to print.\"], \"headings\": [\"Customer is credited at the rate billed\", \"Goes back into stock\", \"Sales are reduced in the accounts\"], \"text\": \"A customer brings goods back.\\nIn Bill History select the invoice they were bought on.\\nPress Return Goods.\\nType the Return qty for each item. Tick Spoiled? if the goods are rotten and cannot be sold again.\\nWrite a reason (optional) and, for a walk-in customer, choose how you refund (Cash, UPI, Bank).\\nPress Save return. You are offered the credit note to print.\\nReturn goods\\nWhat happens:\\n | Good goods | Spoiled goods\\nCustomer is credited at the rate billed | yes | yes\\nGoes back into stock | yes, at the cost it was sold at | no\\nSales are reduced in the accounts | yes | yes\\nThe credit first reduces what the customer still owes on that invoice; anything more stays as credit on their account (shown as Cr in the customer list).\\nYou cannot return more than was billed minus what was already returned.\\nA void invoice cannot have goods returned.\\nA walk-in customer is refunded from the cash drawer or bank.\"}, \"credit-notes\": {\"id\": \"credit-notes\", \"title\": \"Credit notes\", \"number\": \"3.6\", \"chapter\": \"history\", \"chapter_title\": \"Bill history and returns\", \"screen\": \"Bill History\", \"keywords\": [\"credit note\", \"print credit note\", \"reprint\", \"list of returns\", \"credit note pdf\"], \"related\": [\"return-goods\", \"cancel-return\"], \"summary\": \"Bill History → Credit Notes lists the returns made on the selected invoice, each with its number, date and amount.\", \"steps\": [], \"headings\": [], \"text\": \"Bill History → Credit Notes lists the returns made on the selected invoice, each with its number, date and amount.\\nCredit notes for an invoice\\nSelect one and press Print / Save PDF. The credit note shows each item, whether it was good or spoiled, the total in words and how the credit was applied.\\nCredit note\"}, \"cancel-return\": {\"id\": \"cancel-return\", \"title\": \"Cancel a return\", \"number\": \"3.7\", \"chapter\": \"history\", \"chapter_title\": \"Bill history and returns\", \"screen\": \"Bill History, Procurement\", \"keywords\": [\"cancel return\", \"undo return\", \"return by mistake\", \"cancel credit note\", \"cancel debit note\", \"reverse return\"], \"related\": [\"return-goods\", \"purchase-return\", \"credit-notes\"], \"summary\": \"A return entered by mistake can be cancelled.\", \"steps\": [\"**Bill History → Credit Notes**, select the return.\", \"Press **Cancel this return** and confirm.\"], \"headings\": [], \"text\": \"A return entered by mistake can be cancelled.\\nBill History → Credit Notes, select the return.\\nPress Cancel this return and confirm.\\nBillDesk puts back the customer's balance, the invoice's balance due, the stock and the accounts exactly as they were; the credit note stays on file, marked cancelled, and no longer counts anywhere. The whole quantity can be returned again afterwards.\\nA return can be cancelled once only.\\nReturns made by the older program cannot be cancelled here.\\nA supplier return (debit note) is cancelled from Procurement → Returns to Suppliers → Cancel Debit Note (purchase return).\"}, \"orders-list\": {\"id\": \"orders-list\", \"title\": \"Orders and shipments\", \"number\": \"4.1\", \"chapter\": \"orders\", \"chapter_title\": \"Customer orders\", \"screen\": \"Orders\", \"keywords\": [\"orders\", \"order list\", \"pending orders\", \"orders and shipments\", \"Ctrl+O\", \"edit order\", \"cancel order\", \"print order\"], \"related\": [\"order-new\", \"order-convert\", \"order-matrix\"], \"summary\": \"Open File → Ordering System or press Ctrl+O.\", \"steps\": [], \"headings\": [\"+ Create New Order\", \"Edit Order\", \"Convert to Sales Bill\", \"Convert to Purchase Bill\", \"Print / Preview\", \"Cancel Order\"], \"text\": \"Open File → Ordering System or press Ctrl+O.\\nOrders\\nThe cards show total, pending and today's orders.\\nSearch by order number or customer; the Date box filters by order date (pick it from the calendar).\\nOrder date filter\\nButton | What it does\\n+ Create New Order | opens the order form\\nEdit Order | changes an order that is not yet billed\\nConvert to Sales Bill | makes the bill from the order (convert)\\nConvert to Purchase Bill | turns the order into a purchase from a supplier\\nPrint / Preview | prints the order as a challan\\nCancel Order | cancels an order that has not been billed\\nStatus: pending, confirmed, billed, cancelled.\"}, \"order-new\": {\"id\": \"order-new\", \"title\": \"New customer order\", \"number\": \"4.2\", \"chapter\": \"orders\", \"chapter_title\": \"Customer orders\", \"screen\": \"New Order\", \"keywords\": [\"new order\", \"create order\", \"customer order\", \"order form\", \"F3\", \"F5\", \"F8\", \"delivery date\", \"crates\", \"save order\", \"whatsapp order\"], \"related\": [\"order-dates\", \"smart-importer\", \"order-convert\"], \"summary\": \"Press + Create New Order on the Orders screen, or File → New Customer Order.\", \"steps\": [\"Press **F5** and choose the customer.\", \"The cursor goes to the **order date**, then the **delivery date**, then the first empty row of the items.\", \"Enter the items exactly as on a bill (code, quantity, unit, rate).\", \"Enter **Crates out / in** if returnable crates go with the order.\", \"Press **F3** to save (or **F10** to save and print).\"], \"headings\": [\"F3\", \"F5\", \"F8\", \"F10\", \"Esc\"], \"text\": \"Press + Create New Order on the Orders screen, or File → New Customer Order.\\nPress F5 and choose the customer.\\nThe cursor goes to the order date, then the delivery date, then the first empty row of the items.\\nEnter the items exactly as on a bill (code, quantity, unit, rate).\\nEnter Crates out / in if returnable crates go with the order.\\nPress F3 to save (or F10 to save and print).\\nNew order\\nKey | Does\\nF3 | save the order (F2 does the same)\\nF5 | choose the customer, from any box\\nF8 | Smart Importer: read the order from a photo or text\\nF10 | save and print\\nEsc | close the form\\nOrders need a customer from the customer master.\\nThe commission and mandi fee shown at the bottom come from your settings (both are zero unless your administrator sets them).\"}, \"order-dates\": {\"id\": \"order-dates\", \"title\": \"Order and delivery dates\", \"number\": \"4.3\", \"chapter\": \"orders\", \"chapter_title\": \"Customer orders\", \"screen\": \"New Order\", \"keywords\": [\"order date\", \"delivery date\", \"calendar\", \"delivery tomorrow\", \"date before\", \"invalid date\"], \"related\": [\"order-new\", \"dates-everywhere\"], \"summary\": \"Both date boxes drop down the calendar when you click them or reach them with the keyboard.\", \"steps\": [], \"headings\": [], \"text\": \"Both date boxes drop down the calendar when you click them or reach them with the keyboard.\\nDelivery date calendar\\nThe order date cannot be in the future.\\nThe delivery date cannot be before the order date (those days are greyed out in the calendar). It starts as tomorrow.\\nBoth are required; an unusable date turns red and the form tells you which one when you save.\"}, \"smart-importer\": {\"id\": \"smart-importer\", \"title\": \"Smart Importer (photo, scan or text)\", \"number\": \"4.4\", \"chapter\": \"orders\", \"chapter_title\": \"Customer orders\", \"screen\": \"New Order\", \"keywords\": [\"smart importer\", \"F8\", \"ocr\", \"photo\", \"image\", \"scan\", \"tesseract\", \"tamil\", \"translate\", \"whatsapp\", \"handwritten\", \"read order from picture\", \"upload image\"], \"related\": [\"text-importer\", \"order-new\"], \"summary\": \"Customers send orders as photos or messages. F8 on the order form reads them for you.\", \"steps\": [\"Press **F8**.\", \"Press **Upload Image…** and choose the photo or scan (PNG, JPG, BMP, TIFF, WebP).\", \"Wait a few seconds: the picture shows on the left and what was read on the right.\", \"Check the **customer**, the **date** and every row. A row marked **? (choose item)** could not be matched: **double-click** it to pick the item, correct the quantity or unit, or remove the line.\", \"Press **Import to Grid**. The items go into the order; check the quantities, then save with **F3**.\"], \"headings\": [], \"text\": \"Customers send orders as photos or messages. F8 on the order form reads them for you.\\nPress F8.\\nPress Upload Image… and choose the photo or scan (PNG, JPG, BMP, TIFF, WebP).\\nWait a few seconds: the picture shows on the left and what was read on the right.\\nCheck the customer, the date and every row. A row marked ? (choose item) could not be matched: double-click it to pick the item, correct the quantity or unit, or remove the line.\\nPress Import to Grid. The items go into the order; check the quantities, then save with F3.\\nSmart Importer\\nBillDesk reads English and Tamil (and Hindi, Telugu, Kannada, Malayalam) with the built-in Tesseract reader, converts Tamil numerals, translates the item names to English and matches them to your Item Master.\\nThe date and the customer at the top of the note are picked up too.\\nPrinted or typed orders read well. Handwriting is hard for this kind of reader: if nothing can be read the window says so instead of guessing. Use a straight, bright photo, or type the lines with Paste text instead.\\nThe same item twice in one order is added together.\"}, \"text-importer\": {\"id\": \"text-importer\", \"title\": \"Importing order text\", \"number\": \"4.5\", \"chapter\": \"orders\", \"chapter_title\": \"Customer orders\", \"screen\": \"New Order\", \"keywords\": [\"paste text\", \"whatsapp text\", \"text importer\", \"import file\", \"csv\", \"parse\"], \"related\": [\"smart-importer\", \"order-new\"], \"summary\": \"For orders that arrive as text (WhatsApp, SMS): in the Smart Importer choose Paste text instead.\", \"steps\": [\"Paste the message, one item per line: `111 30kg`, `beans 12 kg`, `105 20`.\", \"Press **Parse & Populate** (or **Load File…** for a `.txt` / `.csv`).\"], \"headings\": [], \"text\": \"For orders that arrive as text (WhatsApp, SMS): in the Smart Importer choose Paste text instead.\\nText importer\\nPaste the message, one item per line: 111 30kg, beans 12 kg, 105 20.\\nPress Parse & Populate (or Load File… for a .txt / .csv).\\nItem codes and names both work, in either order with the quantity. Lines that cannot be matched to an item are not guessed: they are listed so you can add them by hand.\"}, \"order-convert\": {\"id\": \"order-convert\", \"title\": \"Convert an order to a bill\", \"number\": \"4.6\", \"chapter\": \"orders\", \"chapter_title\": \"Customer orders\", \"screen\": \"Orders\", \"keywords\": [\"convert order\", \"order to bill\", \"make bill from order\", \"billed\", \"deliver order\", \"purchase from order\"], \"related\": [\"orders-list\", \"new-bill-overview\"], \"summary\": \"The bill is made with the order's items and rates and the order becomes billed, linked to the invoice (shown in Linked Docs). A billed order cannot be converted again. Convert to Purchase Bill does the same for buying the goods from a supplier.\", \"steps\": [\"On **Orders**, select the order.\", \"Press **Convert to Sales Bill**.\"], \"headings\": [], \"text\": \"On Orders, select the order.\\nPress Convert to Sales Bill.\\nThe bill is made with the order's items and rates and the order becomes billed, linked to the invoice (shown in Linked Docs). A billed order cannot be converted again. Convert to Purchase Bill does the same for buying the goods from a supplier.\"}, \"order-matrix\": {\"id\": \"order-matrix\", \"title\": \"Order matrix: how much to buy\", \"number\": \"4.7\", \"chapter\": \"orders\", \"chapter_title\": \"Customer orders\", \"screen\": \"Order Matrix\", \"keywords\": [\"order matrix\", \"consolidation\", \"how much to buy\", \"shortfall\", \"to purchase\", \"items to procure\", \"delivery date\", \"demand\"], \"related\": [\"orders-list\", \"inventory\"], \"summary\": \"Reports → Order Consolidation adds up all pending orders: one row per item, one column per customer, then the total order, current stock and to purchase.\", \"steps\": [], \"headings\": [], \"text\": \"Reports → Order Consolidation adds up all pending orders: one row per item, one column per customer, then the total order, current stock and to purchase.\\nOrder matrix\\nDelivery Date limits the matrix to orders due on that day (pick it from the calendar); All shows every pending order.\\nMatrix for one delivery date\\nShow Stock and Show To Purchase switch those columns; Export to Excel and Print Configuration produce a copy.\"}, \"items\": {\"id\": \"items\", \"title\": \"Items\", \"number\": \"5.1\", \"chapter\": \"masters\", \"chapter_title\": \"Masters: items, customers, suppliers\", \"screen\": \"Item Master\", \"keywords\": [\"items\", \"item master\", \"add item\", \"product\", \"code\", \"alias\", \"rate\", \"unit\", \"stock\", \"F10\", \"new item\", \"edit item\"], \"related\": [\"fixed-rates\", \"field-rules\"], \"summary\": \"Masters → Item Master (F10).\", \"steps\": [\"Press **+ Add Item**.\", \"Fill the form and press **Save Item**. To change an item, double-click it (or **Edit Selected**).\"], \"headings\": [\"Alias (Fast Code)\", \"Item Name\", \"Category / Unit of Measure\", \"Standard Rate (₹)\", \"Initial Stock Quantity\"], \"text\": \"Masters → Item Master (F10).\\nItem Master\\nPress + Add Item.\\nFill the form and press Save Item. To change an item, double-click it (or Edit Selected).\\nAdd item\\nBox | Meaning\\nAlias (Fast Code) | the short code you type when billing (111); no spaces, unique\\nItem Name | up to 80 characters; two items cannot have the same name\\nCategory / Unit of Measure | for grouping; the unit appears on the bill\\nStandard Rate (₹) | the rate used when a customer has no fixed rate (0 or more)\\nInitial Stock Quantity | opening quantity (0 or more)\"}, \"customers\": {\"id\": \"customers\", \"title\": \"Customers\", \"number\": \"5.2\", \"chapter\": \"masters\", \"chapter_title\": \"Masters: items, customers, suppliers\", \"screen\": \"Customer Master\", \"keywords\": [\"customers\", \"customer master\", \"add customer\", \"phone\", \"address\", \"credit limit\", \"balance\", \"advance\", \"Dr\", \"Cr\", \"edit customer\", \"F4\", \"statement\", \"GSTIN\"], \"related\": [\"statement\", \"fixed-rates\", \"field-rules\"], \"summary\": \"Masters → Customer Master (F4).\", \"steps\": [], \"headings\": [], \"text\": \"Masters → Customer Master (F4).\\nCustomer Master\\nBalance shows Dr when the customer owes you and Cr when you hold their advance or a return credit.\\nButtons: + Add Customer, Edit Customer, Statement (the customer's account, bill by bill: statement) and Refresh.\\nAdd customer\\nCustomer Name (ship-to) and DC Company Name (the \\\"bill to\\\") can differ: goods go to one place, the invoice to another company. Invoices and the consolidated report use both.\\nPhone: 7 to 15 digits, up to three numbers separated by commas. Email and GSTIN are optional but checked when given (GSTIN is 15 letters or digits). Names must be unique.\\nCredit limit and Opening balance must be numbers. A customer's balance is changed only by bills, receipts and returns, never by editing the form.\"}, \"suppliers\": {\"id\": \"suppliers\", \"title\": \"Suppliers\", \"number\": \"5.3\", \"chapter\": \"masters\", \"chapter_title\": \"Masters: items, customers, suppliers\", \"screen\": \"Supplier Master\", \"keywords\": [\"suppliers\", \"vendors\", \"supplier master\", \"add supplier\", \"TDS\", \"farmer\", \"F11\", \"edit supplier\"], \"related\": [\"vendor-bill\", \"supplier-payment\"], \"summary\": \"Masters → Supplier Master.\", \"steps\": [], \"headings\": [], \"text\": \"Masters → Supplier Master.\\nSupplier Master\\nAdd supplier\\nName, phone, address and GSTIN (optional) as for customers. If TDS applies to the supplier, set the TDS rate (0 to 100): vendor bills then deduct it automatically.\"}, \"fixed-rates\": {\"id\": \"fixed-rates\", \"title\": \"Fixed contract rates\", \"number\": \"5.4\", \"chapter\": \"masters\", \"chapter_title\": \"Masters: items, customers, suppliers\", \"screen\": \"Fixed Rates\", \"keywords\": [\"fixed rate\", \"contract rate\", \"special price\", \"agreed price\", \"customer price\", \"validity\", \"start date\", \"end date\"], \"related\": [\"items\", \"customers\", \"bill-items\"], \"summary\": \"A customer with an agreed price for an item gets that rate automatically on every bill and order while the contract is valid.\", \"steps\": [\"**Reports → Fixed Rates Report**, press **+ Add Contract Price**.\", \"Enter the **customer ID**, the **item ID or code** and the **rate** (must be above zero).\", \"Pick the **start** and **end dates** from the calendars (the end cannot be before the start).\", \"Press **Save Price**.\"], \"headings\": [], \"text\": \"A customer with an agreed price for an item gets that rate automatically on every bill and order while the contract is valid.\\nFixed rates\\nReports → Fixed Rates Report, press + Add Contract Price.\\nEnter the customer ID, the item ID or code and the rate (must be above zero).\\nPick the start and end dates from the calendars (the end cannot be before the start).\\nPress Save Price.\\nAdd contract price\\nA new price for the same customer and item replaces the old one. The billing hint line says fixed rate for <customer> when one is used.\"}, \"inventory\": {\"id\": \"inventory\", \"title\": \"Inventory: live stock\", \"number\": \"6.1\", \"chapter\": \"stock\", \"chapter_title\": \"Stock and purchases\", \"screen\": \"Inventory\", \"keywords\": [\"inventory\", \"stock\", \"live stock\", \"godown\", \"quantity on hand\", \"stock overview\", \"movements\", \"history\"], \"related\": [\"stock-adjust\", \"waste\", \"crates\"], \"summary\": \"Masters → Inventory & Stock opens the inventory screen on Live Stock Overview (Masters → Waste Management opens the same screen on the waste tab). It has four tabs.\", \"steps\": [], \"headings\": [], \"text\": \"Masters → Inventory & Stock opens the inventory screen on Live Stock Overview (Masters → Waste Management opens the same screen on the waste tab). It has four tabs.\\nLive stock\\nLive Stock Overview: every item with its quantity on hand.\\nStock Movement History: every sale, return, goods receipt, adjustment and waste that changed stock, with date and reference.\\nStock movements\\nProduce Waste Logs and Crate Balances are described next.\\nStock for an item may show 0 if you never entered it: BillDesk then does not block sales. Goods receipts and adjustments start it.\"}, \"stock-adjust\": {\"id\": \"stock-adjust\", \"title\": \"Stock adjustment (after counting)\", \"number\": \"6.2\", \"chapter\": \"stock\", \"chapter_title\": \"Stock and purchases\", \"screen\": \"Inventory\", \"keywords\": [\"stock adjustment\", \"stock count\", \"correct stock\", \"adjust quantity\", \"physical count\", \"add stock\", \"reduce stock\"], \"related\": [\"inventory\", \"waste\"], \"summary\": \"Select the item on Live Stock Overview.\", \"steps\": [\"Select the item on **Live Stock Overview**.\", \"Press **Manual Adjustment (+/-)**.\", \"Type the change (`+5` to add, `-3` to reduce) and the **reason** (required for the audit trail).\", \"Press **Commit Adjustment**.\"], \"headings\": [], \"text\": \"Select the item on Live Stock Overview.\\nPress Manual Adjustment (+/-).\\nType the change (+5 to add, -3 to reduce) and the reason (required for the audit trail).\\nPress Commit Adjustment.\\nStock adjustment\\nThe item can be entered by its id or its code.\\nStock cannot go below zero by adjusting; the quantity must be a number.\"}, \"waste\": {\"id\": \"waste\", \"title\": \"Waste and spoilage\", \"number\": \"6.3\", \"chapter\": \"stock\", \"chapter_title\": \"Stock and purchases\", \"screen\": \"Inventory\", \"keywords\": [\"waste\", \"spoilage\", \"spoiled\", \"rotten\", \"damaged\", \"write off\", \"loss\", \"wastage\"], \"related\": [\"stock-adjust\", \"return-goods\", \"inventory\"], \"summary\": \"Produce that rots or is damaged is recorded as waste, so it reduces stock and shows as a loss in your profit and loss.\", \"steps\": [\"Select the item, press **Record Waste / Spoilage**.\", \"Enter the **quantity**, the **estimated cost rate** and the **reason** (Rotten, Crushed…).\", \"Press **Record Waste**.\"], \"headings\": [], \"text\": \"Produce that rots or is damaged is recorded as waste, so it reduces stock and shows as a loss in your profit and loss.\\nSelect the item, press Record Waste / Spoilage.\\nEnter the quantity, the estimated cost rate and the reason (Rotten, Crushed…).\\nPress Record Waste.\\nRecord waste\\nThe Produce Waste Logs tab lists them:\\nWaste logs\\nIf the stock is tracked you cannot waste more than is on hand. Goods a customer returns as spoiled are handled by Return goods, not here.\"}, \"crates\": {\"id\": \"crates\", \"title\": \"Crates\", \"number\": \"6.4\", \"chapter\": \"stock\", \"chapter_title\": \"Stock and purchases\", \"screen\": \"Inventory\", \"keywords\": [\"crates\", \"jali\", \"returnable\", \"crate balance\", \"crates out\", \"crates in\"], \"related\": [\"order-new\", \"inventory\"], \"summary\": \"Returnable crates that go out with a bill or an order (Crates out) and come back (Crates in) are tracked per customer.\", \"steps\": [], \"headings\": [], \"text\": \"Returnable crates that go out with a bill or an order (Crates out) and come back (Crates in) are tracked per customer.\\nCrate balances\\nVoiding a bill reverses its crates. Crate numbers must be whole numbers, 0 or more.\"}, \"procurement-bills\": {\"id\": \"procurement-bills\", \"title\": \"Procurement and vendor bills\", \"number\": \"6.5\", \"chapter\": \"stock\", \"chapter_title\": \"Stock and purchases\", \"screen\": \"Procurement\", \"keywords\": [\"procurement\", \"purchase\", \"vendor bills\", \"purchase bills\", \"supplier invoice\", \"Ctrl+P\", \"payable\", \"TDS\"], \"related\": [\"vendor-bill\", \"purchase-orders\", \"purchase-return\"], \"summary\": \"File → Purchase system or Ctrl+P.\", \"steps\": [], \"headings\": [], \"text\": \"File → Purchase system or Ctrl+P.\\nVendor purchase bills\\nFour tabs: Vendor Purchase Bills, Purchase Orders, Goods Receipt Notes and Returns to Suppliers. The bills tab lists each supplier invoice with subtotal, TDS, net payable, balance due and status.\"}, \"vendor-bill\": {\"id\": \"vendor-bill\", \"title\": \"Enter a vendor bill\", \"number\": \"6.6\", \"chapter\": \"stock\", \"chapter_title\": \"Stock and purchases\", \"screen\": \"Procurement\", \"keywords\": [\"vendor bill\", \"supplier bill\", \"enter purchase\", \"purchase invoice\", \"TDS\", \"buy\", \"payable\", \"duplicate invoice\"], \"related\": [\"supplier-payment\", \"purchase-return\", \"suppliers\"], \"summary\": \"+ Enter Vendor Bill.\", \"steps\": [\"**+ Enter Vendor Bill**.\", \"Type the **Supplier ID**, the supplier's **invoice number**, the **item** (id or code), the **quantity** and the **rate**.\", \"Press **Record Bill**.\"], \"headings\": [], \"text\": \"+ Enter Vendor Bill.\\nType the Supplier ID, the supplier's invoice number, the item (id or code), the quantity and the rate.\\nPress Record Bill.\\nEnter a vendor bill\\nTDS is worked out from the supplier's rate and what you owe is the net payable.\\nThe same supplier invoice number cannot be entered twice (BillDesk tells you the purchase it already belongs to).\\nQuantity and rate must be above zero and the item must exist.\"}, \"purchase-orders\": {\"id\": \"purchase-orders\", \"title\": \"Purchase orders and goods receipts\", \"number\": \"6.7\", \"chapter\": \"stock\", \"chapter_title\": \"Stock and purchases\", \"screen\": \"Procurement\", \"keywords\": [\"purchase order\", \"PO\", \"goods receipt\", \"GRN\", \"received\", \"three way match\", \"receive goods\"], \"related\": [\"vendor-bill\", \"inventory\"], \"summary\": \"A vendor bill linked to a PO and a GRN is checked three ways (rates against the PO, quantities against what was received) and mismatches are flagged.\", \"steps\": [], \"headings\": [], \"text\": \"Purchase Orders: what you asked a supplier to send.\\nPurchase orders\\nGoods Receipt Notes (GRN): what actually arrived against a PO; receiving adds to stock.\\nGoods receipts\\nA vendor bill linked to a PO and a GRN is checked three ways (rates against the PO, quantities against what was received) and mismatches are flagged.\"}, \"purchase-return\": {\"id\": \"purchase-return\", \"title\": \"Return goods to a supplier (debit note)\", \"number\": \"6.8\", \"chapter\": \"stock\", \"chapter_title\": \"Stock and purchases\", \"screen\": \"Procurement\", \"keywords\": [\"return to supplier\", \"debit note\", \"purchase return\", \"send back\", \"supplier credit\", \"cancel debit note\"], \"related\": [\"cancel-return\", \"vendor-bill\", \"supplier-payment\"], \"summary\": \"Procurement → Vendor Purchase Bills, select the bill the goods came on.\", \"steps\": [\"**Procurement → Vendor Purchase Bills**, select the bill the goods came on.\", \"Press **↩ Return Goods to Supplier**.\", \"Type the quantity of each item, the reason, and press **Save return**. You can print the **debit note**.\"], \"headings\": [], \"text\": \"Procurement → Vendor Purchase Bills, select the bill the goods came on.\\nPress ↩ Return Goods to Supplier.\\nType the quantity of each item, the reason, and press Save return. You can print the debit note.\\nReturn to supplier\\nThe value is at the bill's rate. If TDS was deducted, the same share comes back, so what you owe falls by the net amount.\\nIt reduces what you still owe on that bill first; any excess is credit with the supplier.\\nStock goes down only if the goods had been received into stock (the bill has a goods receipt).\\nThe Returns to Suppliers tab lists them; Print Debit Note and Cancel Debit Note are there.\\nReturns to suppliers\\nDebit note\"}, \"accounting-home\": {\"id\": \"accounting-home\", \"title\": \"Accounting dashboard\", \"number\": \"7.1\", \"chapter\": \"accounts\", \"chapter_title\": \"Money and accounts\", \"screen\": \"Finance\", \"keywords\": [\"accounting\", \"finance\", \"accounts dashboard\", \"Ctrl+D\", \"receivables total\", \"payables\", \"bank balance\", \"net profit\"], \"related\": [\"receivables\", \"payables\", \"banking\", \"statements\"], \"summary\": \"Accounts → Accounting Dashboard (Ctrl+D).\", \"steps\": [], \"headings\": [], \"text\": \"Accounts → Accounting Dashboard (Ctrl+D).\\nAccounting dashboard\\nThe cards show total receivables, payables, bank balance and net profit; the quick buttons open the other tabs: Accounts Receivable (AR), Accounts Payable (AP), Banking & BRS and General Ledger.\"}, \"receivables\": {\"id\": \"receivables\", \"title\": \"Receivables: who owes you\", \"number\": \"7.2\", \"chapter\": \"accounts\", \"chapter_title\": \"Money and accounts\", \"screen\": \"Accounts Receivables\", \"keywords\": [\"receivables\", \"who owes\", \"customer balance\", \"aging\", \"overdue\", \"outstanding\", \"AR\", \"collect money\"], \"related\": [\"receipt\", \"statement\", \"history-filters\"], \"summary\": \"Accounts → Accounts Receivables.\", \"steps\": [], \"headings\": [], \"text\": \"Accounts → Accounts Receivables.\\nReceivables\\nEach customer who owes money, split by age: Current, 1–30, 31–60, 61–90 and 90+ days. The total here equals the customer balances, the ledger and the dashboard card. Customers who have paid in advance are not subtracted: they are shown in the line above as Advances held.\\n+ Record Customer Receipt records a payment (receipt).\\nCustomer Statement opens the selected customer's account (statement).\"}, \"receipt\": {\"id\": \"receipt\", \"title\": \"Customer receipts\", \"number\": \"7.3\", \"chapter\": \"accounts\", \"chapter_title\": \"Money and accounts\", \"screen\": \"Accounts Receivables\", \"keywords\": [\"receipt\", \"record payment\", \"customer payment\", \"received money\", \"collect\", \"UPI\", \"cheque\", \"NEFT\", \"advance\", \"FIFO\", \"pay later\"], \"related\": [\"statement\", \"receivables\", \"bill-payment\"], \"summary\": \"When a customer pays later (not at the counter):\", \"steps\": [\"**Accounts Receivables → + Record Customer Receipt**.\", \"Enter the **Customer ID**, the **amount** and choose the **payment mode** from the list.\", \"Leave **Invoice No** blank to settle the oldest bills first (FIFO), or type one invoice number.\", \"Add the UTR / cheque reference and press **Save Receipt**.\"], \"headings\": [], \"text\": \"When a customer pays later (not at the counter):\\nAccounts Receivables → + Record Customer Receipt.\\nEnter the Customer ID, the amount and choose the payment mode from the list.\\nLeave Invoice No blank to settle the oldest bills first (FIFO), or type one invoice number.\\nAdd the UTR / cheque reference and press Save Receipt.\\nCustomer receipt\\nThe customer's balance goes down and the matching bills are marked paid. An amount larger than what they owe becomes their advance.\"}, \"statement\": {\"id\": \"statement\", \"title\": \"Customer statement\", \"number\": \"7.4\", \"chapter\": \"accounts\", \"chapter_title\": \"Money and accounts\", \"screen\": \"Customer Master, Receivables\", \"keywords\": [\"statement\", \"customer account\", \"ledger\", \"bill by bill\", \"what does a customer owe\", \"running balance\", \"export\", \"csv\"], \"related\": [\"receipt\", \"customers\", \"receivables\"], \"summary\": \"From Customer Master → Statement or Accounts Receivables → Customer Statement.\", \"steps\": [], \"headings\": [], \"text\": \"From Customer Master → Statement or Accounts Receivables → Customer Statement.\\nCustomer statement\\nLists invoices (billed), receipts, returns and adjustments with a running balance, ending on the customer's balance.\\nFrom / To (calendar boxes) show one period; the earlier balance is brought forward.\\nExport CSV saves it for Excel.\\n\\\"Adjustment\\\" appears when the customer paid something that was never recorded here.\"}, \"payables\": {\"id\": \"payables\", \"title\": \"Payables and supplier payments\", \"number\": \"7.5\", \"chapter\": \"accounts\", \"chapter_title\": \"Money and accounts\", \"screen\": \"Accounts Payables\", \"keywords\": [\"payables\", \"what do i owe\", \"supplier balance\", \"AP\", \"owe suppliers\"], \"related\": [\"supplier-payment\", \"vendor-bill\"], \"summary\": \"Accounts → Accounts Payables lists your vendor bills with net payable, balance due and status.\", \"steps\": [], \"headings\": [], \"text\": \"Accounts → Accounts Payables lists your vendor bills with net payable, balance due and status.\\nPayables\"}, \"supplier-payment\": {\"id\": \"supplier-payment\", \"title\": \"Pay a supplier\", \"number\": \"7.6\", \"chapter\": \"accounts\", \"chapter_title\": \"Money and accounts\", \"screen\": \"Accounts Payables\", \"keywords\": [\"pay supplier\", \"disburse\", \"supplier payment\", \"NEFT\", \"RTGS\", \"cheque\", \"part payment\"], \"related\": [\"payables\", \"vendor-bill\"], \"summary\": \"You cannot pay more than the bill's balance due.\", \"steps\": [\"**Accounts Payables → + Disburse Supplier Payment**.\", \"Enter the **Purchase ID**, the **Supplier ID**, the **amount** and the **payment mode**.\", \"Add the UTR and press **Disburse Payment**.\"], \"headings\": [], \"text\": \"Accounts Payables → + Disburse Supplier Payment.\\nEnter the Purchase ID, the Supplier ID, the amount and the payment mode.\\nAdd the UTR and press Disburse Payment.\\nSupplier payment\\nYou cannot pay more than the bill's balance due.\"}, \"banking\": {\"id\": \"banking\", \"title\": \"Banking and reconciliation\", \"number\": \"7.7\", \"chapter\": \"accounts\", \"chapter_title\": \"Money and accounts\", \"screen\": \"BRS\", \"keywords\": [\"bank\", \"banking\", \"BRS\", \"bank reconciliation\", \"bank account\", \"IFSC\", \"statement\", \"match\", \"auto reconcile\"], \"related\": [\"journal\", \"accounting-home\"], \"summary\": \"Accounts → BRS.\", \"steps\": [\"**+ Add Bank Account** and fill in the form.\"], \"headings\": [\"Bank name\", \"Account number\", \"IFSC\", \"Account type\", \"Opening balance\"], \"text\": \"Accounts → BRS.\\nBanking\\n+ Add Bank Account and fill in the form.\\nAdd bank account\\nBox | Rule\\nBank name | required\\nAccount number | 6–20 digits, registered once\\nIFSC | optional; 4 letters, a 0, 6 letters/digits (SBIN0001234)\\nAccount type | Current, Savings or OD\\nOpening balance | a number\\n⚡ Auto-Reconciliation matches bank statement lines with your records by reference and amount; View BRS Statement shows the reconciliation.\"}, \"journal\": {\"id\": \"journal\", \"title\": \"General ledger and manual journal\", \"number\": \"7.8\", \"chapter\": \"accounts\", \"chapter_title\": \"Money and accounts\", \"screen\": \"Finance\", \"keywords\": [\"general ledger\", \"journal\", \"double entry\", \"post journal\", \"manual entry\", \"debit\", \"credit\", \"account code\", \"opening balance\"], \"related\": [\"statements\", \"accounting-home\"], \"summary\": \"The General Ledger tab lists every accounting entry. BillDesk posts them automatically for bills, receipts, purchases, payments, returns and waste; use a manual journal only for what has no screen (for example opening capital).\", \"steps\": [\"**+ Post Manual Journal**.\", \"Enter a **reference** (`ADJ-001`), the **debit account code** and amount, the **credit account code** and amount.\", \"Press **Post Entry**.\"], \"headings\": [], \"text\": \"The General Ledger tab lists every accounting entry. BillDesk posts them automatically for bills, receipts, purchases, payments, returns and waste; use a manual journal only for what has no screen (for example opening capital).\\nGeneral ledger\\n+ Post Manual Journal.\\nEnter a reference (ADJ-001), the debit account code and amount, the credit account code and amount.\\nPress Post Entry.\\nManual journal\\nThe two accounts must be different and the amounts must be equal; otherwise the form tells you before anything is posted.\"}, \"statements\": {\"id\": \"statements\", \"title\": \"Trial balance, profit and loss, balance sheet\", \"number\": \"7.9\", \"chapter\": \"accounts\", \"chapter_title\": \"Money and accounts\", \"screen\": \"Finance\", \"keywords\": [\"trial balance\", \"profit and loss\", \"P&L\", \"balance sheet\", \"financial statements\", \"net profit\", \"assets\", \"liabilities\"], \"related\": [\"journal\", \"integrity\"], \"summary\": \"Accounts → Trial Balance / Profit & Loss / Balance Sheet each open a report window.\", \"steps\": [], \"headings\": [], \"text\": \"Accounts → Trial Balance / Profit & Loss / Balance Sheet each open a report window.\\nTrial balance\\nProfit and loss\\nBalance sheet\\nTrial balance: every account's debits and credits; the totals must be equal.\\nProfit and loss: sales (net of returns), cost of goods sold, waste and expenses, giving the profit.\\nBalance sheet: what you own, owe and the owner's capital.\"}, \"integrity\": {\"id\": \"integrity\", \"title\": \"Integrity check\", \"number\": \"7.10\", \"chapter\": \"accounts\", \"chapter_title\": \"Money and accounts\", \"screen\": \"Integrity Check\", \"keywords\": [\"integrity\", \"check\", \"books balance\", \"audit\", \"errors in data\", \"reconcile\", \"health check\", \"accounts add up\"], \"related\": [\"statements\", \"backup\"], \"summary\": \"Accounts → Integrity Check → Run Check reads your data (it changes nothing) and tests that the books are consistent.\", \"steps\": [], \"headings\": [], \"text\": \"Accounts → Integrity Check → Run Check reads your data (it changes nothing) and tests that the books are consistent.\\nIntegrity check\\nEach line is OK, Info, Warning or Fail: journals balance, receivables equal customer balances, bills add up, statuses agree with balances, invoice numbers are unique, stock is not negative, and more. Open a line to see exactly which records. Save Report… keeps a copy. Run it before closing the month and after restoring a backup.\"}, \"daybook\": {\"id\": \"daybook\", \"title\": \"Daybook\", \"number\": \"8.1\", \"chapter\": \"reports\", \"chapter_title\": \"Reports\", \"screen\": \"Reports\", \"keywords\": [\"daybook\", \"day book\", \"cash book\", \"money in\", \"money out\", \"today's transactions\", \"all transactions\", \"cashbook\"], \"related\": [\"item-wise\", \"customer-wise\", \"report-dates\"], \"summary\": \"Reports → Daybook: every event in date order: sales, returns, purchases, receipts and supplier payments, with Billed, Bought, Money in and Money out and a total line.\", \"steps\": [], \"headings\": [], \"text\": \"Reports → Daybook: every event in date order: sales, returns, purchases, receipts and supplier payments, with Billed, Bought, Money in and Money out and a total line.\\nDaybook\\nChoose the period with the From / To boxes (dates in reports), or Today / This month. Export CSV saves it.\"}, \"item-wise\": {\"id\": \"item-wise\", \"title\": \"Item-wise sales\", \"number\": \"8.2\", \"chapter\": \"reports\", \"chapter_title\": \"Reports\", \"screen\": \"Reports\", \"keywords\": [\"item wise\", \"best selling\", \"top items\", \"sales by item\", \"quantity sold\", \"average rate\", \"which item sells\"], \"related\": [\"daybook\", \"customer-wise\"], \"summary\": \"Reports → Item-wise Sales: for each item the bills it was on, quantity sold, quantity returned, net quantity, average rate and net amount, best sellers first.\", \"steps\": [], \"headings\": [], \"text\": \"Reports → Item-wise Sales: for each item the bills it was on, quantity sold, quantity returned, net quantity, average rate and net amount, best sellers first.\\nItem-wise sales\"}, \"customer-wise\": {\"id\": \"customer-wise\", \"title\": \"Customer-wise sales\", \"number\": \"8.3\", \"chapter\": \"reports\", \"chapter_title\": \"Reports\", \"screen\": \"Reports\", \"keywords\": [\"customer wise\", \"sales by customer\", \"top customers\", \"who buys most\", \"returns by customer\"], \"related\": [\"daybook\", \"item-wise\", \"statement\"], \"summary\": \"Reports → Customer-wise Sales: bills, billed, returned, net sales and what each customer owes now.\", \"steps\": [], \"headings\": [], \"text\": \"Reports → Customer-wise Sales: bills, billed, returned, net sales and what each customer owes now.\\nCustomer-wise sales\"}, \"report-dates\": {\"id\": \"report-dates\", \"title\": \"Dates in reports\", \"number\": \"8.4\", \"chapter\": \"reports\", \"chapter_title\": \"Reports\", \"screen\": \"Reports\", \"keywords\": [\"report dates\", \"from to\", \"period\", \"this month\", \"today\", \"export csv\"], \"related\": [\"daybook\", \"dates-everywhere\"], \"summary\": \"All three reports share the From / To boxes.\", \"steps\": [], \"headings\": [], \"text\": \"All three reports share the From / To boxes.\\nReport calendar\\nPick a day or type it; the To date cannot be before the From date.\\nLeave a box empty for \\\"no limit\\\".\\nThe reports always net out returns, and cancelled returns are ignored.\"}, \"consolidated\": {\"id\": \"consolidated\", \"title\": \"Consolidated bills\", \"number\": \"8.5\", \"chapter\": \"reports\", \"chapter_title\": \"Reports\", \"screen\": \"Consolidated Billing\", \"keywords\": [\"consolidated\", \"statement of bills\", \"bill to\", \"multiple bills one invoice\", \"month end statement\", \"group bills\", \"PDF\"], \"related\": [\"customers\", \"history\"], \"summary\": \"Reports → Bills Consolidated Report: one statement of all bills for a \\\"bill to\\\" company over a period, grouped by delivery address.\", \"steps\": [\"Press the search icon and choose the **Bill To** company.\", \"Pick **From** and **To** (they start at the 1st of this month and today).\", \"Press **Generate**, then save or print the PDF.\"], \"headings\": [], \"text\": \"Reports → Bills Consolidated Report: one statement of all bills for a \\\"bill to\\\" company over a period, grouped by delivery address.\\nPress the search icon and choose the Bill To company.\\nPick From and To (they start at the 1st of this month and today).\\nPress Generate, then save or print the PDF.\\nConsolidated bills\"}, \"cash-drawer\": {\"id\": \"cash-drawer\", \"title\": \"Cash drawer\", \"number\": \"9.1\", \"chapter\": \"admin\", \"chapter_title\": \"Administration\", \"screen\": \"Handover & Settlement\", \"keywords\": [\"cash drawer\", \"cashier\", \"session\", \"float\", \"z report\", \"handover\", \"close drawer\", \"counted cash\", \"shortage\", \"excess\"], \"related\": [\"users\", \"bill-payment\"], \"summary\": \"Accounts → Handover & Settlement.\", \"steps\": [\"At the start of the day press **Open Drawer Session**, type the **opening float** and press **Start Session**.\"], \"headings\": [], \"text\": \"Accounts → Handover & Settlement.\\nCash drawer\\nAt the start of the day press Open Drawer Session, type the opening float and press Start Session.\\nOpen the drawer\\nWork as usual: cash receipts are added to the expected cash.\\nAt the end press Close Drawer & Generate Report, type the cash you counted and a note for any difference.\\nBillDesk shows the Z-report: expected, counted and short / excess.\\nAmounts must be numbers (0 or more). You can have only one open session.\"}, \"users\": {\"id\": \"users\", \"title\": \"Users and roles\", \"number\": \"9.2\", \"chapter\": \"admin\", \"chapter_title\": \"Administration\", \"screen\": \"User Management\", \"keywords\": [\"users\", \"add user\", \"roles\", \"permissions\", \"access\", \"password\", \"cashier\", \"manager\", \"admin\", \"accountant\", \"who can do what\"], \"related\": [\"signing-in\", \"company\"], \"summary\": \"Settings → User Management, tab Users & Access.\", \"steps\": [\"**+ Add Operator User**.\", \"Type a **username** (3–30 letters, digits, dot, dash or underscore, unique), a **password** (at least 6 characters, not the same as the username), choose the **role** from the list, add email and phone (optional).\", \"Press **Create User**.\"], \"headings\": [\"user (cashier)\", \"manager\", \"admin\"], \"text\": \"Settings → User Management, tab Users & Access.\\nUsers\\n+ Add Operator User.\\nType a username (3–30 letters, digits, dot, dash or underscore, unique), a password (at least 6 characters, not the same as the username), choose the role from the list, add email and phone (optional).\\nPress Create User.\\nAdd user\\nThe role decides which menus a person sees:\\nRole | Sees\\nuser (cashier) | File (billing, orders, purchases), Bills History, Bills Consolidated Report, Order Consolidation, Dashboard, Help\\nmanager | all of the above, plus Masters (items, customers, suppliers, inventory, waste, fixed rates), Accounts, Daybook, item-wise and customer-wise sales, Profit & Loss, Balance Sheet\\nadmin | everything, including Settings (users, company, database, audit logs)\\nA menu never offers a screen you cannot open. A forgotten password is reset by the administrator with scripts\\\\reset_password.py.\"}, \"company\": {\"id\": \"company\", \"title\": \"Company configuration\", \"number\": \"9.3\", \"chapter\": \"admin\", \"chapter_title\": \"Administration\", \"screen\": \"Company Settings\", \"keywords\": [\"company\", \"company settings\", \"logo\", \"address\", \"phone\", \"email\", \"GSTIN\", \"terms and conditions\", \"signatory\", \"invoice header\", \"default company\", \"add company\"], \"related\": [\"bill-print\", \"users\"], \"summary\": \"Settings → Company Settings: your business details as they appear on every invoice.\", \"steps\": [], \"headings\": [], \"text\": \"Settings → Company Settings: your business details as they appear on every invoice.\\nCompany configuration\\n+ Add Company and the ✏ edit icon open the form; the trash icon deletes (the default company and the last company cannot be deleted).\\nSearch, Export and Import (a CSV with a name column) manage many companies.\\nEdit company\\nName, address, phones, email, GSTIN (optional), terms and conditions, signatory title, a logo (any picture; it is shrunk) and use as default company. Changes appear on the next invoice you print.\"}, \"backup\": {\"id\": \"backup\", \"title\": \"Backup and database\", \"number\": \"9.4\", \"chapter\": \"admin\", \"chapter_title\": \"Administration\", \"screen\": \"Database Settings\", \"keywords\": [\"backup\", \"restore\", \"back up\", \"data safe\", \"database\", \"connection\", \"copy data\", \"lost data\", \"recover\", \"log file\", \"backup folder\"], \"related\": [\"integrity\", \"faq\"], \"summary\": \"Settings → Database Settings (titled Backup & Database).\", \"steps\": [], \"headings\": [], \"text\": \"Settings → Database Settings (titled Backup & Database).\\nBackup and database\\nA backup is taken automatically once a day while BillDesk is used, and each one is checked after it is written.\\nBack Up Now makes one immediately; Open Backup Folder shows the files; Open Log File helps your IT person.\\nTo restore, your IT person runs python scripts\\\\restore_backup.py <backup.zip> --target-db <new name>; then run the Integrity check.\\nWhere your data is kept normally needs no change. Test Connection checks the database address.\\nCopy the backup folder to a pen drive or cloud drive regularly: a backup on the same computer does not help if the computer is lost.\"}, \"audit-logs\": {\"id\": \"audit-logs\", \"title\": \"System audit logs\", \"number\": \"9.5\", \"chapter\": \"admin\", \"chapter_title\": \"Administration\", \"screen\": \"System Audit Logs\", \"keywords\": [\"audit\", \"audit log\", \"who did what\", \"activity\", \"history of changes\", \"user activity\"], \"related\": [\"users\"], \"summary\": \"Settings → System Audit Logs answers who did what, and when. Nothing on it can be changed. It has two lists:\", \"steps\": [\"**Bill changes:** every time a bill was made, voided, or had goods returned (or a return cancelled): the time, the user, the invoice, the action and, where something changed, what it was before and after.\", \"**Sign-ins and activity:** who signed in and when.\"], \"headings\": [], \"text\": \"Settings → System Audit Logs answers who did what, and when. Nothing on it can be changed. It has two lists:\\nBill changes: every time a bill was made, voided, or had goods returned (or a return cancelled): the time, the user, the invoice, the action and, where something changed, what it was before and after.\\nSign-ins and activity: who signed in and when.\\nAudit logs: bill changes\\nAudit logs: sign-ins\\nFrom / To (calendar boxes), Action (All, or one kind), and Search (any word: a user, an invoice number) narrow the list; press Show or Enter.\\nExport CSV saves what is on screen.\\nThe newest 2,000 entries are listed; use the dates to look further back.\"}, \"shortcuts\": {\"id\": \"shortcuts\", \"title\": \"Keyboard shortcuts\", \"number\": \"10.1\", \"chapter\": \"reference\", \"chapter_title\": \"Reference\", \"screen\": \"All\", \"keywords\": [\"shortcut\", \"shortcuts\", \"keys\", \"keyboard\", \"hotkey\", \"F1\", \"F2\", \"F3\", \"F4\", \"F5\", \"F6\", \"F7\", \"F8\", \"F9\", \"F10\", \"F11\", \"F12\", \"Esc\", \"Ctrl\"], \"related\": [\"main-window\", \"bill-items\"], \"summary\": \"The strip at the bottom of each screen lists the keys that work there.\", \"steps\": [], \"headings\": [\"F1\", \"F2\", \"F3\", \"F4\", \"F5\", \"F6\", \"F7\", \"F8\", \"F9\", \"F10\", \"F11\", \"F12\", \"Esc\", \"Ctrl+N\", \"Ctrl+O\", \"Ctrl+P\", \"Ctrl+D\"], \"text\": \"The strip at the bottom of each screen lists the keys that work there.\\nKey | Anywhere | On New Bill | On New Order\\nF1 | Dashboard (search on Bill History) |  | \\nF2 | New Bill | save and take payment | save the order\\nF3 | Item Master | save and print | save the order\\nF4 | Customer Master |  | \\nF5 | refresh the screen | choose customer | choose customer\\nF6 | Bills History | park the bill | \\nF7 |  | parked bills | \\nF8 |  |  | Smart Importer\\nF9 | Help for this screen |  | \\nF10 | Item Master |  | save and print\\nF11 | Supplier Master |  | \\nF12 | Log out |  | \\nEsc | close a dialog | close a dialog | close the form\\nCtrl+N | New Customer Order |  | \\nCtrl+O | Orders |  | \\nCtrl+P | Procurement |  | \\nCtrl+D | Accounting Dashboard |  | \\nIn lists and dialogs: ↑ / ↓ move, Enter chooses, Esc closes. In the grids: Enter moves to the next box. In any date box: ↓ opens the calendar.\"}, \"dates-everywhere\": {\"id\": \"dates-everywhere\", \"title\": \"Dates and the calendar\", \"number\": \"10.2\", \"chapter\": \"reference\", \"chapter_title\": \"Reference\", \"screen\": \"All\", \"keywords\": [\"date\", \"calendar\", \"date box\", \"date format\", \"dd/mm/yyyy\", \"future date\", \"from to\", \"red date\", \"invalid date\"], \"related\": [\"bill-date\", \"order-dates\", \"report-dates\"], \"summary\": \"Every date box in BillDesk works the same way:\", \"steps\": [], \"headings\": [\"Bill date\", \"Order date\", \"Delivery date\", \"Fixed rate start / end\", \"Report and statement From / To\", \"Consolidated report From / To\", \"Filters (Bill History, Orders)\"], \"text\": \"Every date box in BillDesk works the same way:\\nClick it (or move to it with the keyboard): a calendar drops down. Click a day, or use the arrow keys (Page Up / Down change month) and Enter; Today jumps to today.\\nYou can also type a date: 13/10/2026, 13-10-26, 2026-10-13. Press Enter to accept it.\\nText turns red when the date cannot be used; the form tells you why when you save.\\nDays that are not allowed are greyed in the calendar.\\nWhere | Rule\\nBill date | required; not in the future\\nOrder date | required; not in the future\\nDelivery date | required; not before the order date\\nFixed rate start / end | required; end not before start\\nReport and statement From / To | optional; To not before From\\nConsolidated report From / To | required; To not before From\\nFilters (Bill History, Orders) | optional; part of a date (09/2026) is fine\"}, \"field-rules\": {\"id\": \"field-rules\", \"title\": \"What each form accepts\", \"number\": \"10.3\", \"chapter\": \"reference\", \"chapter_title\": \"Reference\", \"screen\": \"All forms\", \"keywords\": [\"validation\", \"rules\", \"field rules\", \"what can i type\", \"limits\", \"invalid\", \"required\", \"phone\", \"email\", \"GSTIN\", \"IFSC\", \"negative\", \"number\", \"decimal\"], \"related\": [\"messages\", \"dates-everywhere\"], \"summary\": \"BillDesk checks what you type and says what is wrong instead of saving bad data.\", \"steps\": [], \"headings\": [\"Quantity, rate, amount\", \"Item name\", \"Item code\", \"Customer / supplier name\", \"Phone\", \"Email\", \"GSTIN\", \"Credit limit, opening balance\", \"TDS rate\", \"Username\", \"Password\", \"Payment mode\", \"Vendor invoice number\", \"Bank account\", \"Cash drawer amounts\", \"Crates\", \"Journal entry\"], \"text\": \"BillDesk checks what you type and says what is wrong instead of saving bad data.\\nField | Rule\\nQuantity, rate, amount | a real number; quantity above zero; rate zero or more; not absurdly large\\nItem name | required, up to 80 characters, not already used\\nItem code | no spaces, up to 20 characters, unique\\nCustomer / supplier name | required, up to 80 characters, not already used\\nPhone | 7–15 digits; up to three numbers separated by comma or slash\\nEmail | like name@shop.com\\nGSTIN | optional; 15 letters/digits\\nCredit limit, opening balance | a number (credit limit 0 or more)\\nTDS rate | 0 to 100\\nUsername | 3–30 letters, digits, . - _\\nPassword | at least 6 characters, at most 72 bytes, not the username\\nPayment mode | Cash, UPI, Cheque, NEFT, RTGS, NEFT/RTGS, Bank, Card (and Credit/Due on a bill)\\nVendor invoice number | required, up to 40 characters, once per supplier\\nBank account | 6–20 digits; IFSC like SBIN0001234; type Current, Savings or OD\\nCash drawer amounts | 0 or more\\nCrates | whole numbers, 0 or more\\nJournal entry | reference required; two different accounts; debit equals credit\"}, \"messages\": {\"id\": \"messages\", \"title\": \"Messages and what to do\", \"number\": \"10.4\", \"chapter\": \"reference\", \"chapter_title\": \"Reference\", \"screen\": \"All\", \"keywords\": [\"error\", \"message\", \"problem\", \"not working\", \"cannot\", \"why\", \"failed\", \"warning\", \"troubleshoot\", \"fix\", \"wrong\"], \"related\": [\"field-rules\", \"faq\", \"backup\"], \"summary\": \"Find the words you see on the screen in the first column; the next columns say why it happened and what to do.\", \"steps\": [], \"headings\": [\"\\\"Choose a customer from the customer master (F5)\\\"\", \"\\\"The bill date cannot be in the future\\\"\", \"\\\"… is not a valid date. Use DD/MM/YYYY\\\"\", \"\\\"The To date cannot be before …\\\"\", \"\\\"Amount received (…) is more than the bill total\\\"\", \"\\\"Item … not found\\\" / \\\"No item matches\\\"\", \"\\\"… must be a number\\\"\", \"\\\"An item / customer named … already exists\\\"\", \"\\\"Vendor invoice … is already recorded as PUR-…\\\"\", \"\\\"Payment amount … exceeds purchase balance due\\\"\", \"\\\"Waste quantity … exceeds stock on hand\\\"\", \"\\\"Invoice … has goods returned against it, so it cannot be voided\\\"\", \"\\\"only n can still be returned\\\"\", \"\\\"Return … was made by the older BillDesk and cannot be cancelled here\\\"\", \"\\\"The entry does not balance\\\"\", \"\\\"You already have an open cashier session\\\"\", \"\\\"Access Denied\\\"\", \"\\\"Tesseract OCR engine is not installed\\\"\", \"\\\"The text in this image could not be read reliably\\\"\", \"Integrity check shows Warning\", \"Cannot start: database error\"], \"text\": \"Find the words you see on the screen in the first column; the next columns say why it happened and what to do.\\nWhat you see | Why | What to do\\n\\\"Choose a customer from the customer master (F5)\\\" | an order needs a real customer | press F5, choose one; add the customer first if missing\\n\\\"The bill date cannot be in the future\\\" | the date box holds a later day | pick today or an earlier day\\n\\\"… is not a valid date. Use DD/MM/YYYY\\\" | text in a date box can't be read | pick from the calendar\\n\\\"The To date cannot be before …\\\" | To is earlier than From | change one of them\\n\\\"Amount received (…) is more than the bill total\\\" | more entered than the bill | type the bill amount or less\\n\\\"Item … not found\\\" / \\\"No item matches\\\" | wrong code | check the code in the Item Master\\n\\\"… must be a number\\\" | letters or symbols in a number box | type digits only (decimals allowed)\\n\\\"An item / customer named … already exists\\\" | the name is taken | use another name, or edit the existing one\\n\\\"Vendor invoice … is already recorded as PUR-…\\\" | the same supplier invoice was entered before | do not enter it again\\n\\\"Payment amount … exceeds purchase balance due\\\" | paying a supplier more than owed | pay the balance due or less\\n\\\"Waste quantity … exceeds stock on hand\\\" | more than the tracked stock | check the quantity\\n\\\"Invoice … has goods returned against it, so it cannot be voided\\\" | a return exists | cancel the return first, then void\\n\\\"only n can still be returned\\\" | more than billed minus earlier returns | reduce the return quantity\\n\\\"Return … was made by the older BillDesk and cannot be cancelled here\\\" | old returns lack detail | leave it; it is already reflected in the balance\\n\\\"The entry does not balance\\\" | journal debit ≠ credit | make the two amounts equal\\n\\\"You already have an open cashier session\\\" | a drawer is open | close it first\\n\\\"Access Denied\\\" | your role does not include that screen | ask the administrator\\n\\\"Tesseract OCR engine is not installed\\\" | the image reader is missing | use Paste text instead, or ask IT to reinstall\\n\\\"The text in this image could not be read reliably\\\" | handwriting or a blurry photo | take a straighter, brighter photo or type the lines\\nIntegrity check shows Warning | an older-data difference | open the line; ask your accountant if unsure\\nCannot start: database error | the database did not start | restart the computer, then BillDesk; call IT if it repeats\"}, \"faq\": {\"id\": \"faq\", \"title\": \"Frequently asked questions\", \"number\": \"10.5\", \"chapter\": \"reference\", \"chapter_title\": \"Reference\", \"screen\": \"All\", \"keywords\": [\"faq\", \"questions\", \"common questions\", \"help\", \"tips\"], \"related\": [\"messages\", \"start-here\"], \"summary\": \"Short answers to the questions people ask most; follow the link in the last column for the full steps.\", \"steps\": [], \"headings\": [\"Can I change the date of a bill after saving?\", \"How do I sell to someone who is not a customer?\", \"Why did the bill number have a different date?\", \"A customer paid half. What now?\", \"A customer paid later in cash. How do I record it?\", \"How do I see what one customer owes, bill by bill?\", \"What is \\\"Cr\\\" next to a balance?\", \"What is the difference between void and return?\", \"I returned goods by mistake.\", \"Does BillDesk handle GST?\", \"Can two people bill at once?\", \"How often should I back up?\", \"Can it read my customer's handwritten list?\", \"Stock shows 0 but I can still sell. Is that wrong?\", \"I forgot my password.\", \"Where do I change my shop name and address on invoices?\"], \"text\": \"Short answers to the questions people ask most; follow the link in the last column for the full steps.\\nQuestion | Answer | More\\nCan I change the date of a bill after saving? | No. Void the bill and make it again with the right date. | Void\\nHow do I sell to someone who is not a customer? | Leave Cash as the customer and take the full payment. | Customer\\nWhy did the bill number have a different date? | The invoice number follows the bill date you chose. | Bill date\\nA customer paid half. What now? | Enter the amount received; the rest stays on their account. | Payment\\nA customer paid later in cash. How do I record it? | Use Record Customer Receipt. | Receipt\\nHow do I see what one customer owes, bill by bill? | Open the customer's Statement. | Statement\\nWhat is \\\"Cr\\\" next to a balance? | You hold the customer's money (advance or return credit). | Customers\\nWhat is the difference between void and return? | Void cancels the whole bill; return takes back some goods and keeps the bill. | Return\\nI returned goods by mistake. | Cancel the return. | Cancel return\\nDoes BillDesk handle GST? | No: fruit and vegetables are exempt, so no GST is calculated. | What it does\\nCan two people bill at once? | Yes on different computers using the same database; park a bill to serve two customers at one counter. | Park\\nHow often should I back up? | BillDesk backs up daily; also copy the backup folder to another drive weekly. | Backup\\nCan it read my customer's handwritten list? | Not reliably. Printed or typed lists work; otherwise type or paste. | Smart Importer\\nStock shows 0 but I can still sell. Is that wrong? | No. Stock only tracks what you entered through purchases or adjustments. | Inventory\\nI forgot my password. | Ask the administrator; they reset it. | Signing in\\nWhere do I change my shop name and address on invoices? | Settings → Company Settings. | Company\"}, \"glossary\": {\"id\": \"glossary\", \"title\": \"Glossary\", \"number\": \"10.6\", \"chapter\": \"reference\", \"chapter_title\": \"Reference\", \"screen\": \"All\", \"keywords\": [\"glossary\", \"meaning\", \"words\", \"terms\", \"DC\", \"GRN\", \"PO\", \"BRS\", \"TDS\", \"FIFO\", \"advance\", \"credit note\", \"debit note\", \"challan\"], \"related\": [\"start-here\"], \"summary\": \"The words used in BillDesk and in this guide, in plain language.\", \"steps\": [], \"headings\": [\"Bill / Invoice\", \"Challan (DC)\", \"Void\", \"Credit note\", \"Debit note\", \"Dr / Cr\", \"Advance\", \"Receivables\", \"Payables\", \"PO / GRN\", \"TDS\", \"FIFO\", \"BRS\", \"Trial balance\", \"Float\", \"Legacy bill\", \"Parked bill\"], \"text\": \"The words used in BillDesk and in this guide, in plain language.\\nWord | Meaning\\nBill / Invoice | the document you give the customer for goods sold\\nChallan (DC) | the same bill without prices, for delivery\\nVoid | cancel a whole bill, keeping its record\\nCredit note | document for goods a customer returned (their balance goes down)\\nDebit note | document for goods you send back to a supplier\\nDr / Cr | Dr: the customer owes you; Cr: you hold their money\\nAdvance | money received before a bill or more than was owed\\nReceivables | money customers owe you\\nPayables | money you owe suppliers\\nPO / GRN | purchase order / goods receipt note (what actually arrived)\\nTDS | tax deducted from a supplier's payment\\nFIFO | oldest bill settled first\\nBRS | bank reconciliation statement\\nTrial balance | list of all accounts; debits must equal credits\\nFloat | cash put in the drawer at the start of the day\\nLegacy bill | a bill made by the older program (no per-bill payment record)\\nParked bill | a bill put on hold\"}, \"about-guide\": {\"id\": \"about-guide\", \"title\": \"About this guide\", \"number\": \"10.7\", \"chapter\": \"reference\", \"chapter_title\": \"Reference\", \"screen\": \"Help\", \"keywords\": [\"about\", \"version\", \"rebuild guide\", \"documentation\", \"update guide\"], \"related\": [\"start-here\"], \"summary\": \"This guide is written once in Docs/BillDesk_Native_User_Guide.md and turned into the Word, PDF and Jupyter notebook editions and the in-program Help by python scripts/build_user_guide.py. The screenshots come from python scripts/user_guide/capture_screenshots.py running on a demo database (see scripts/user_guide/README.md), so they can be redone whenever a screen changes.\", \"steps\": [], \"headings\": [], \"text\": \"This guide is written once in Docs/BillDesk_Native_User_Guide.md and turned into the Word, PDF and Jupyter notebook editions and the in-program Help by python scripts/build_user_guide.py. The screenshots come from python scripts/user_guide/capture_screenshots.py running on a demo database (see scripts/user_guide/README.md), so they can be redone whenever a screen changes.\"}}}")
_help = HelpService(data=_DATA)

def ask(question, how_many=5):
    """Print the best topics for a question, with the first steps."""
    from IPython.display import Markdown, display
    hits = _help.search(question, limit=how_many)
    if not hits:
        display(Markdown('Nothing matched. Try fewer or simpler words, e.g. **return**, **backup**, **date**.\n\nIdeas: ' + '; '.join(_help.suggestions(6))))
        return
    out = [f'### Results for "{question}"']
    for n, h in enumerate(hits, 1):
        t = h.topic
        out.append(f'**{n}. [{t["number"]} {t["title"]}](#{t["id"]})**  _{t["chapter_title"]}_')
        if h.why: out.append(f'> {h.why}')
        if h.snippet: out.append(h.snippet)
        if n <= 3 and h.steps: out.append('Quick steps:\n' + '\n'.join(f'{i}. {s}' for i, s in enumerate(h.steps[:4], 1)))
    display(Markdown('\n\n'.join(out)))

ask('how do I return goods')
```
